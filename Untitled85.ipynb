```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import yaml

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

warnings.filterwarnings("ignore")


# =============================================================================
# 2. TENSORFLOW IMPORT
# =============================================================================

try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense, Dropout, Input
    from tensorflow.keras.utils import to_categorical

    TENSORFLOW_AVAILABLE = True

    print("TensorFlow version:", tf.__version__)

except ImportError:
    TENSORFLOW_AVAILABLE = False

    print("\nWARNING:")
    print("TensorFlow is not installed.")
    print("Install it using:")
    print("pip install tensorflow")


# =============================================================================
# 3. FILE PATHS
# =============================================================================

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Government Scheme Fund Allocation Trend Analysis"

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_256_AU_3172_C.csv"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "government_scheme_model.pkl"
)

H5_PATH = os.path.join(
    BASE_DIR,
    "government_scheme_model.h5"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "metadata.json"
)


# Create directory if it does not exist
os.makedirs(BASE_DIR, exist_ok=True)


print("=" * 80)
print("AI-BASED GOVERNMENT SCHEME FUND ALLOCATION TREND ANALYSIS")
print("=" * 80)

print("\nDataset Path:")
print(CSV_PATH)

print("\nOutput Directory:")
print(BASE_DIR)


# =============================================================================
# 4. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n\n"
        "Please verify that the CSV file exists at this location."
    )


# =============================================================================
# 5. LOAD DATASET
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASET")
print("=" * 80)

df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)

print("\nDataset loaded successfully.")

print("\nDataset Shape:")
print(df.shape)

print("\nOriginal Columns:")

for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {repr(col)}")


print("\nFirst 5 rows:")
print(df.head())


# =============================================================================
# 6. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned Columns:")

for col in df.columns:
    print("-", col)


# =============================================================================
# 7. REMOVE COMPLETELY EMPTY ROWS/COLUMNS
# =============================================================================

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)

df = df.reset_index(drop=True)


print("\nShape after removing empty rows/columns:")
print(df.shape)


# =============================================================================
# 8. IDENTIFY YEAR COLUMNS
# =============================================================================

def normalize_year_name(column_name):

    name = str(column_name).lower()

    replacements = {
        "–": "-",
        "—": "-",
        "_": "-",
        "/": "-"
    }

    for old, new in replacements.items():
        name = name.replace(old, new)

    name = name.replace(" ", "")

    return name


year_patterns = {
    "2018-19": [
        "2018-19",
        "201819"
    ],

    "2019-20": [
        "2019-20",
        "201920"
    ],

    "2020-21": [
        "2020-21",
        "202021"
    ]
}


detected_year_columns = {}


for column in df.columns:

    normalized = normalize_year_name(column)

    for standard_year, patterns in year_patterns.items():

        for pattern in patterns:

            if pattern in normalized:

                detected_year_columns[standard_year] = column
                break


print("\nDetected year columns:")

for year, column in detected_year_columns.items():
    print(year, "->", column)


# =============================================================================
# 9. FALLBACK: FIND NUMERIC COLUMNS
# =============================================================================

def clean_numeric_series(series):

    cleaned = (
        series.astype(str)
        .str.strip()
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.replace("Rs.", "", regex=False)
        .str.replace("Rs", "", regex=False)
    )

    cleaned = cleaned.replace(
        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],
        np.nan
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


# If all 3 expected years were not detected,
# automatically find columns that are mostly numeric.

if len(detected_year_columns) < 3:

    print(
        "\nCould not identify all financial-year columns "
        "from their names."
    )

    print(
        "Attempting automatic numeric-column detection..."
    )

    numeric_candidates = []

    for column in df.columns:

        converted = clean_numeric_series(
            df[column]
        )

        valid_ratio = converted.notna().mean()

        if valid_ratio >= 0.50:

            numeric_candidates.append(column)

    print("\nNumeric candidates:")
    print(numeric_candidates)

    if len(numeric_candidates) < 3:

        raise ValueError(
            "Unable to find at least three numeric "
            "financial-year columns."
        )

    # Use the last three numeric columns
    selected_columns = numeric_candidates[-3:]

    detected_year_columns = {
        "2018-19": selected_columns[0],
        "2019-20": selected_columns[1],
        "2020-21": selected_columns[2]
    }


# =============================================================================
# 10. CREATE CLEAN DATAFRAME
# =============================================================================

clean_df = pd.DataFrame()


# Find scheme/name column
year_original_columns = list(
    detected_year_columns.values()
)

non_year_columns = [
    col
    for col in df.columns
    if col not in year_original_columns
]


if len(non_year_columns) > 0:

    scheme_column = non_year_columns[0]

else:

    scheme_column = None


if scheme_column is not None:

    clean_df["Scheme"] = (
        df[scheme_column]
        .astype(str)
        .str.strip()
    )

else:

    clean_df["Scheme"] = [
        f"Scheme_{i + 1}"
        for i in range(len(df))
    ]


# Convert year columns
for year in [
    "2018-19",
    "2019-20",
    "2020-21"
]:

    original_column = detected_year_columns[year]

    clean_df[year] = clean_numeric_series(
        df[original_column]
    )


# =============================================================================
# 11. HANDLE MISSING VALUES
# =============================================================================

year_columns = [
    "2018-19",
    "2019-20",
    "2020-21"
]


# Remove rows where all three values are missing
clean_df = clean_df.dropna(
    subset=year_columns,
    how="all"
)


# Fill individual missing values with row mean
row_mean = clean_df[
    year_columns
].mean(axis=1)


for column in year_columns:

    clean_df[column] = (
        clean_df[column]
        .fillna(row_mean)
    )


# Remaining missing values -> column median
for column in year_columns:

    clean_df[column] = (
        clean_df[column]
        .fillna(
            clean_df[column].median()
        )
    )


clean_df = clean_df.reset_index(drop=True)


print("\nClean Dataset:")
print(clean_df)


print("\nClean Dataset Shape:")
print(clean_df.shape)


# =============================================================================
# 12. FEATURE ENGINEERING
# =============================================================================

print("\n" + "=" * 80)
print("FEATURE ENGINEERING")
print("=" * 80)


EPSILON = 1e-9


# -------------------------------------------------------------------------
# Total Allocation
# -------------------------------------------------------------------------

clean_df["Total_Allocation"] = (
    clean_df[
        year_columns
    ].sum(axis=1)
)


# -------------------------------------------------------------------------
# Average Allocation
# -------------------------------------------------------------------------

clean_df["Average_Allocation"] = (
    clean_df[
        year_columns
    ].mean(axis=1)
)


# -------------------------------------------------------------------------
# Minimum Allocation
# -------------------------------------------------------------------------

clean_df["Minimum_Allocation"] = (
    clean_df[
        year_columns
    ].min(axis=1)
)


# -------------------------------------------------------------------------
# Maximum Allocation
# -------------------------------------------------------------------------

clean_df["Maximum_Allocation"] = (
    clean_df[
        year_columns
    ].max(axis=1)
)


# -------------------------------------------------------------------------
# Standard Deviation
# -------------------------------------------------------------------------

clean_df["Std_Allocation"] = (
    clean_df[
        year_columns
    ].std(axis=1)
)


# -------------------------------------------------------------------------
# Absolute Growth
# -------------------------------------------------------------------------

clean_df["Growth_2018_2019"] = (
    clean_df["2019-20"]
    -
    clean_df["2018-19"]
)


clean_df["Growth_2019_2020"] = (
    clean_df["2020-21"]
    -
    clean_df["2019-20"]
)


clean_df["Overall_Growth"] = (
    clean_df["2020-21"]
    -
    clean_df["2018-19"]
)


# -------------------------------------------------------------------------
# Percentage Growth
# -------------------------------------------------------------------------

clean_df["Growth_Percent_2018_2019"] = np.where(

    np.abs(clean_df["2018-19"]) > EPSILON,

    (
        clean_df["2019-20"]
        -
        clean_df["2018-19"]
    )
    /
    np.abs(clean_df["2018-19"])
    * 100,

    0
)


clean_df["Growth_Percent_2019_2020"] = np.where(

    np.abs(clean_df["2019-20"]) > EPSILON,

    (
        clean_df["2020-21"]
        -
        clean_df["2019-20"]
    )
    /
    np.abs(clean_df["2019-20"])
    * 100,

    0
)


clean_df["Overall_Growth_Percent"] = np.where(

    np.abs(clean_df["2018-19"]) > EPSILON,

    (
        clean_df["2020-21"]
        -
        clean_df["2018-19"]
    )
    /
    np.abs(clean_df["2018-19"])
    * 100,

    0
)


# =============================================================================
# 13. CALCULATE TREND SLOPE
# =============================================================================

def calculate_slope(row):

    x = np.array(
        [0, 1, 2],
        dtype=float
    )

    y = np.array(
        [
            row["2018-19"],
            row["2019-20"],
            row["2020-21"]
        ],
        dtype=float
    )

    slope = np.polyfit(
        x,
        y,
        1
    )[0]

    return float(slope)


clean_df["Trend_Slope"] = clean_df.apply(
    calculate_slope,
    axis=1
)


# =============================================================================
# 14. VOLATILITY
# =============================================================================

clean_df["Coefficient_of_Variation"] = np.where(

    np.abs(
        clean_df["Average_Allocation"]
    ) > EPSILON,

    clean_df["Std_Allocation"]
    /
    np.abs(
        clean_df["Average_Allocation"]
    ),

    0
)


# =============================================================================
# 15. CREATE TREND CLASSIFICATION TARGET
# =============================================================================

# Threshold used to prevent very tiny changes from
# automatically being classified as increasing/decreasing.

STABLE_THRESHOLD_PERCENT = 5.0


def classify_trend(row):

    y1 = row["2018-19"]
    y2 = row["2019-20"]
    y3 = row["2020-21"]

    overall_percent = row[
        "Overall_Growth_Percent"
    ]

    # Stable
    if abs(overall_percent) <= STABLE_THRESHOLD_PERCENT:

        return "Stable"

    # Increasing
    elif y1 < y2 < y3:

        return "Increasing"

    # Decreasing
    elif y1 > y2 > y3:

        return "Decreasing"

    # Everything else
    else:

        return "Fluctuating"


clean_df["Trend_Class"] = clean_df.apply(
    classify_trend,
    axis=1
)


print("\nTrend Classification:")

print(
    clean_df[
        [
            "Scheme",
            "2018-19",
            "2019-20",
            "2020-21",
            "Overall_Growth_Percent",
            "Trend_Slope",
            "Trend_Class"
        ]
    ]
)


print("\nClass Distribution:")

print(
    clean_df[
        "Trend_Class"
    ].value_counts()
)


# =============================================================================
# 16. FEATURE COLUMNS
# =============================================================================

feature_columns = [

    "2018-19",
    "2019-20",
    "2020-21",

    "Total_Allocation",
    "Average_Allocation",
    "Minimum_Allocation",
    "Maximum_Allocation",
    "Std_Allocation",

    "Growth_2018_2019",
    "Growth_2019_2020",
    "Overall_Growth",

    "Growth_Percent_2018_2019",
    "Growth_Percent_2019_2020",
    "Overall_Growth_Percent",

    "Trend_Slope",
    "Coefficient_of_Variation"
]


X = clean_df[
    feature_columns
].copy()


y = clean_df[
    "Trend_Class"
].copy()


# Replace infinity
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


# Fill remaining missing values
X = X.fillna(0)


print("\nNumber of Features:")
print(len(feature_columns))


print("\nFeatures:")

for feature in feature_columns:
    print("-", feature)


# =============================================================================
# 17. LABEL ENCODING
# =============================================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(
    y
)


print("\nEncoded Classes:")

for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "->",
        class_name
    )


num_classes = len(
    label_encoder.classes_
)


# =============================================================================
# 18. TRAIN RANDOM FOREST MODEL
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING RANDOM FOREST MODEL")
print("=" * 80)


# With only a few rows, train/test splitting is not reliable.
# The model is trained on the complete dataset as a demonstration.

rf_pipeline = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                class_weight="balanced"
            )
        )
    ]
)


rf_pipeline.fit(
    X,
    y_encoded
)


rf_predictions = rf_pipeline.predict(
    X
)


rf_training_accuracy = accuracy_score(
    y_encoded,
    rf_predictions
)


print(
    "\nRandom Forest Training Accuracy:",
    round(
        rf_training_accuracy * 100,
        2
    ),
    "%"
)


# =============================================================================
# 19. CREATE PKL MODEL PACKAGE
# =============================================================================

model_package = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "model_type":
        "RandomForestClassifier",

    "model":
        rf_pipeline,

    "label_encoder":
        label_encoder,

    "feature_columns":
        feature_columns,

    "year_columns":
        year_columns,

    "stable_threshold_percent":
        STABLE_THRESHOLD_PERCENT,

    "training_accuracy":
        float(
            rf_training_accuracy
        ),

    "classes":
        label_encoder.classes_.tolist(),

    "dataset_rows":
        int(
            len(clean_df)
        ),

    "dataset_columns":
        int(
            len(clean_df.columns)
        )
}


# =============================================================================
# 20. SAVE PKL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PKL MODEL")
print("=" * 80)


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        model_package,
        file
    )


print("\nPKL saved successfully:")
print(PKL_PATH)


# =============================================================================
# 21. TRAIN NEURAL NETWORK FOR H5 FILE
# =============================================================================

print("\n" + "=" * 80)
print("CREATING H5 MODEL")
print("=" * 80)


h5_saved = False
nn_training_accuracy = None


if TENSORFLOW_AVAILABLE:

    # ---------------------------------------------------------
    # Scale features
    # ---------------------------------------------------------

    nn_scaler = StandardScaler()

    X_scaled = nn_scaler.fit_transform(
        X
    )


    # ---------------------------------------------------------
    # Build model depending on number of classes
    # ---------------------------------------------------------

    if num_classes >= 2:

        y_categorical = to_categorical(
            y_encoded,
            num_classes=num_classes
        )


        model = Sequential(
            [
                Input(
                    shape=(
                        X_scaled.shape[1],
                    )
                ),

                Dense(
                    32,
                    activation="relu"
                ),

                Dropout(
                    0.20
                ),

                Dense(
                    16,
                    activation="relu"
                ),

                Dense(
                    num_classes,
                    activation="softmax"
                )
            ]
        )


        model.compile(
            optimizer="adam",
            loss="categorical_crossentropy",
            metrics=["accuracy"]
        )


        print("\nTraining Neural Network...")


        history = model.fit(
            X_scaled,
            y_categorical,
            epochs=150,
            batch_size=2,
            verbose=0
        )


        # ---------------------------------------------------------
        # Evaluate model
        # ---------------------------------------------------------

        loss, nn_training_accuracy = model.evaluate(
            X_scaled,
            y_categorical,
            verbose=0
        )


        print(
            "\nNeural Network Training Accuracy:",
            round(
                nn_training_accuracy * 100,
                2
            ),
            "%"
        )


        # ---------------------------------------------------------
        # Save H5
        # ---------------------------------------------------------

        model.save(
            H5_PATH
        )


        h5_saved = True


        print("\nH5 saved successfully:")
        print(H5_PATH)


    else:

        print(
            "\nOnly one trend class was detected."
        )

        print(
            "A multi-class neural network cannot "
            "meaningfully be trained."
        )


else:

    print(
        "\nTensorFlow unavailable. "
        "H5 model was not generated."
    )


# =============================================================================
# 22. SAVE SCALER INFORMATION FOR H5 MODEL
# =============================================================================

if TENSORFLOW_AVAILABLE and h5_saved:

    scaler_info = {

        "mean":
            nn_scaler.mean_.tolist(),

        "scale":
            nn_scaler.scale_.tolist(),

        "var":
            nn_scaler.var_.tolist()
    }

else:

    scaler_info = {}


# =============================================================================
# 23. CREATE YAML CONFIGURATION
# =============================================================================

print("\n" + "=" * 80)
print("CREATING YAML CONFIGURATION")
print("=" * 80)


config = {

    "project": {

        "name":
            "AI-Based Government Scheme Fund Allocation Trend Analysis",

        "task":
            "Government Scheme Funding Trend Classification",

        "version":
            "1.0"
    },


    "dataset": {

        "file":
            "RS_Session_256_AU_3172_C.csv",

        "path":
            CSV_PATH,

        "rows":
            int(
                len(clean_df)
            ),

        "financial_years":
            year_columns
    },


    "trend_classification": {

        "classes":
            label_encoder.classes_.tolist(),

        "stable_threshold_percent":
            STABLE_THRESHOLD_PERCENT
    },


    "features": feature_columns,


    "random_forest": {

        "n_estimators":
            200,

        "random_state":
            42,

        "class_weight":
            "balanced",

        "training_accuracy":
            float(
                rf_training_accuracy
            )
    },


    "neural_network": {

        "h5_generated":
            bool(
                h5_saved
            ),

        "epochs":
            150,

        "batch_size":
            2,

        "training_accuracy":
            (
                float(
                    nn_training_accuracy
                )
                if nn_training_accuracy
                is not None
                else None
            )
    },


    "output_files": {

        "pkl":
            os.path.basename(
                PKL_PATH
            ),

        "h5":
            os.path.basename(
                H5_PATH
            ),

        "yaml":
            os.path.basename(
                YAML_PATH
            ),

        "json":
            os.path.basename(
                JSON_PATH
            )
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print("\nYAML saved successfully:")
print(YAML_PATH)


# =============================================================================
# 24. CREATE JSON METADATA
# =============================================================================

print("\n" + "=" * 80)
print("CREATING JSON METADATA")
print("=" * 80)


metadata = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "description":
        (
            "Analysis and classification of government "
            "scheme fund allocation patterns across "
            "multiple financial years."
        ),

    "created_at":
        datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),

    "dataset": {

        "filename":
            os.path.basename(
                CSV_PATH
            ),

        "rows":
            int(
                len(clean_df)
            ),

        "original_columns":
            [
                str(col)
                for col in df.columns
            ],

        "financial_years":
            year_columns
    },


    "features":
        feature_columns,


    "target":
        "Trend_Class",


    "classes":
        label_encoder.classes_.tolist(),


    "class_mapping": {

        str(index):
            str(class_name)

        for index, class_name
        in enumerate(
            label_encoder.classes_
        )
    },


    "class_distribution": {

        str(key):
            int(value)

        for key, value
        in clean_df[
            "Trend_Class"
        ].value_counts().items()
    },


    "model": {

        "primary_model":
            "RandomForestClassifier",

        "training_accuracy":
            float(
                rf_training_accuracy
            ),

        "h5_model_generated":
            bool(
                h5_saved
            ),

        "neural_network_training_accuracy":
            (
                float(
                    nn_training_accuracy
                )
                if nn_training_accuracy
                is not None
                else None
            )
    },


    "h5_scaler":
        scaler_info,


    "trend_rules": {

        "increasing":
            (
                "2018-19 < 2019-20 < 2020-21 "
                "and overall change exceeds "
                "stable threshold"
            ),

        "decreasing":
            (
                "2018-19 > 2019-20 > 2020-21 "
                "and overall change exceeds "
                "stable threshold"
            ),

        "stable":
            (
                f"Absolute overall growth percentage "
                f"<= {STABLE_THRESHOLD_PERCENT}%"
            ),

        "fluctuating":
            (
                "Values do not continuously increase "
                "or decrease"
            )
    },


    "output_files": {

        "government_scheme_model.pkl":
            "Random Forest model, preprocessing pipeline, "
            "features and label encoder",

        "government_scheme_model.h5":
            "Keras neural network model",

        "config.yaml":
            "Project and model configuration",

        "metadata.json":
            "Dataset and model metadata"
    },


    "warning":
        (
            "The source dataset contains very few records. "
            "Training accuracy should not be interpreted as "
            "evidence of real-world predictive performance."
        )
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )


print("\nJSON saved successfully:")
print(JSON_PATH)


# =============================================================================
# 25. VERIFY OUTPUT FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING OUTPUT FILES")
print("=" * 80)


output_files = [

    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]


if h5_saved:
    output_files.append(
        H5_PATH
    )


for file_path in output_files:

    if os.path.exists(file_path):

        file_size = os.path.getsize(
            file_path
        )

        print(
            "\n[OK]",
            os.path.basename(file_path)
        )

        print(
            "Size:",
            round(
                file_size / 1024,
                2
            ),
            "KB"
        )

    else:

        print(
            "\n[NOT FOUND]",
            file_path
        )


# =============================================================================
# 26. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("PROJECT SUMMARY")
print("=" * 80)


print(
    "\nNumber of Schemes:",
    len(clean_df)
)


print(
    "Number of Features:",
    len(feature_columns)
)


print(
    "Trend Classes:",
    label_encoder.classes_.tolist()
)


print(
    "Random Forest Training Accuracy:",
    f"{rf_training_accuracy * 100:.2f}%"
)


if nn_training_accuracy is not None:

    print(
        "Neural Network Training Accuracy:",
        f"{nn_training_accuracy * 100:.2f}%"
    )


print("\nGenerated Files:")

print(
    "1.",
    PKL_PATH
)

if h5_saved:

    print(
        "2.",
        H5_PATH
    )

else:

    print(
        "2. H5 not generated because TensorFlow "
        "was unavailable or insufficient classes existed."
    )


print(
    "3.",
    YAML_PATH
)

print(
    "4.",
    JSON_PATH
)


print("\n" + "=" * 80)
print("PROCESS COMPLETED SUCCESSFULLY")
print("=" * 80)
```
