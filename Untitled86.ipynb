```python
# =============================================================================
# AI-BASED GOVERNMENT SCHEME FUND ALLOCATION TREND ANALYSIS
#
# OUTPUTS:
# 1. accuracy_graph.png
# 2. heatmap.png
# 3. comparison_graph.png
# 4. result.csv
# 5. result_graph.png
# 6. prediction.csv
# 7. prediction_graph.png
#
# All graphs are both SHOWN and SAVED.
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
from sklearn.model_selection import LeaveOneOut

warnings.filterwarnings("ignore")


# =============================================================================
# 2. PATH CONFIGURATION
# =============================================================================

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Government Scheme Fund Allocation Trend Analysis"

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_256_AU_3172_C.csv"
)

os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# Output paths

ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)


print("=" * 90)
print("AI-BASED GOVERNMENT SCHEME FUND ALLOCATION TREND ANALYSIS")
print("=" * 90)

print("\nDataset:")
print(CSV_PATH)

print("\nOutput Directory:")
print(BASE_DIR)


# =============================================================================
# 3. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n\n"
        "Please verify the path."
    )


# =============================================================================
# 4. LOAD DATASET
# =============================================================================

print("\n" + "=" * 90)
print("LOADING DATASET")
print("=" * 90)

df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)


print("\nDataset loaded successfully.")

print("\nShape:")
print(df.shape)

print("\nOriginal Columns:")

for i, col in enumerate(
    df.columns,
    start=1
):
    print(
        f"{i}. {repr(col)}"
    )


print("\nFirst Rows:")
print(df.head())


# =============================================================================
# 5. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)


df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)

df = df.reset_index(
    drop=True
)


print("\nCleaned Columns:")

for col in df.columns:
    print("-", col)


# =============================================================================
# 6. NUMERIC CLEANING FUNCTION
# =============================================================================

def clean_numeric_series(series):

    cleaned = (
        series
        .astype(str)
        .str.strip()
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.replace("Rs.", "", regex=False)
        .str.replace("Rs", "", regex=False)
    )

    cleaned = cleaned.replace(
        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],
        np.nan
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


# =============================================================================
# 7. IDENTIFY FINANCIAL YEAR COLUMNS
# =============================================================================

def normalize_column_name(name):

    name = str(name).lower()

    name = (
        name
        .replace("–", "-")
        .replace("—", "-")
        .replace("_", "-")
        .replace("/", "-")
        .replace(" ", "")
    )

    return name


year_patterns = {

    "2018-19": [
        "2018-19",
        "201819"
    ],

    "2019-20": [
        "2019-20",
        "201920"
    ],

    "2020-21": [
        "2020-21",
        "202021"
    ]
}


detected_year_columns = {}


for column in df.columns:

    normalized = normalize_column_name(
        column
    )

    for year, patterns in year_patterns.items():

        for pattern in patterns:

            if pattern in normalized:

                detected_year_columns[
                    year
                ] = column

                break


print("\nDetected Financial Year Columns:")

for year, column in detected_year_columns.items():

    print(
        year,
        "->",
        column
    )


# =============================================================================
# 8. FALLBACK NUMERIC COLUMN DETECTION
# =============================================================================

if len(detected_year_columns) < 3:

    print(
        "\nFinancial-year names were not fully detected."
    )

    print(
        "Searching for numeric columns..."
    )


    numeric_candidates = []


    for column in df.columns:

        converted = clean_numeric_series(
            df[column]
        )

        valid_ratio = (
            converted
            .notna()
            .mean()
        )

        if valid_ratio >= 0.50:

            numeric_candidates.append(
                column
            )


    print(
        "\nNumeric Candidates:",
        numeric_candidates
    )


    if len(numeric_candidates) < 3:

        raise ValueError(
            "Unable to detect three financial-year columns."
        )


    selected = numeric_candidates[-3:]


    detected_year_columns = {

        "2018-19":
            selected[0],

        "2019-20":
            selected[1],

        "2020-21":
            selected[2]
    }


# =============================================================================
# 9. IDENTIFY SCHEME COLUMN
# =============================================================================

year_original_columns = list(
    detected_year_columns.values()
)


non_year_columns = [

    col
    for col in df.columns

    if col not in year_original_columns
]


if len(non_year_columns) > 0:

    scheme_column = non_year_columns[0]

else:

    scheme_column = None


# =============================================================================
# 10. CREATE CLEAN DATAFRAME
# =============================================================================

data = pd.DataFrame()


if scheme_column is not None:

    data["Scheme"] = (
        df[scheme_column]
        .astype(str)
        .str.strip()
    )

else:

    data["Scheme"] = [

        f"Scheme_{i + 1}"

        for i in range(
            len(df)
        )
    ]


for year in [

    "2018-19",
    "2019-20",
    "2020-21"

]:

    original_column = (
        detected_year_columns[
            year
        ]
    )

    data[year] = clean_numeric_series(
        df[original_column]
    )


# =============================================================================
# 11. REMOVE INVALID ROWS
# =============================================================================

YEAR_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21"

]


data = data.dropna(
    subset=YEAR_COLUMNS,
    how="all"
)


# =============================================================================
# 12. HANDLE MISSING VALUES
# =============================================================================

row_means = (
    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


for column in YEAR_COLUMNS:

    data[column] = (
        data[column]
        .fillna(
            row_means
        )
    )


for column in YEAR_COLUMNS:

    median_value = (
        data[column]
        .median()
    )

    data[column] = (
        data[column]
        .fillna(
            median_value
        )
    )


data = data.reset_index(
    drop=True
)


print("\nClean Dataset:")
print(data)


print(
    "\nNumber of Schemes:",
    len(data)
)


# =============================================================================
# 13. FEATURE ENGINEERING
# =============================================================================

print("\n" + "=" * 90)
print("FEATURE ENGINEERING")
print("=" * 90)


EPSILON = 1e-9


# Total
data["Total_Allocation"] = (

    data[
        YEAR_COLUMNS
    ]
    .sum(axis=1)
)


# Average
data["Average_Allocation"] = (

    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


# Minimum
data["Minimum_Allocation"] = (

    data[
        YEAR_COLUMNS
    ]
    .min(axis=1)
)


# Maximum
data["Maximum_Allocation"] = (

    data[
        YEAR_COLUMNS
    ]
    .max(axis=1)
)


# Standard deviation
data["Std_Allocation"] = (

    data[
        YEAR_COLUMNS
    ]
    .std(axis=1)
)


# =============================================================================
# 14. ABSOLUTE GROWTH
# =============================================================================

data["Growth_2018_2019"] = (

    data["2019-20"]
    -
    data["2018-19"]
)


data["Growth_2019_2020"] = (

    data["2020-21"]
    -
    data["2019-20"]
)


data["Overall_Growth"] = (

    data["2020-21"]
    -
    data["2018-19"]
)


# =============================================================================
# 15. PERCENTAGE GROWTH
# =============================================================================

data[
    "Growth_Percent_2018_2019"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2019-20"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


data[
    "Growth_Percent_2019_2020"
] = np.where(

    np.abs(
        data["2019-20"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2019-20"]
        )
        /
        np.abs(
            data["2019-20"]
        )
    )
    * 100,

    0
)


data[
    "Overall_Growth_Percent"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


# =============================================================================
# 16. TREND SLOPE
# =============================================================================

def calculate_slope(row):

    x = np.array(
        [0, 1, 2],
        dtype=float
    )

    y = np.array(
        [
            row["2018-19"],
            row["2019-20"],
            row["2020-21"]
        ],
        dtype=float
    )

    return float(
        np.polyfit(
            x,
            y,
            1
        )[0]
    )


data["Trend_Slope"] = data.apply(
    calculate_slope,
    axis=1
)


# =============================================================================
# 17. COEFFICIENT OF VARIATION
# =============================================================================

data[
    "Coefficient_of_Variation"
] = np.where(

    np.abs(
        data[
            "Average_Allocation"
        ]
    ) > EPSILON,

    (
        data[
            "Std_Allocation"
        ]
        /
        np.abs(
            data[
                "Average_Allocation"
            ]
        )
    ),

    0
)


# =============================================================================
# 18. TREND CLASSIFICATION
# =============================================================================

STABLE_THRESHOLD = 5.0


def classify_trend(row):

    y1 = row["2018-19"]
    y2 = row["2019-20"]
    y3 = row["2020-21"]

    growth = row[
        "Overall_Growth_Percent"
    ]


    if abs(growth) <= STABLE_THRESHOLD:

        return "Stable"


    elif y1 < y2 < y3:

        return "Increasing"


    elif y1 > y2 > y3:

        return "Decreasing"


    else:

        return "Fluctuating"


data[
    "Trend_Class"
] = data.apply(
    classify_trend,
    axis=1
)


print("\nTrend Classes:")

print(
    data[
        [
            "Scheme",
            "2018-19",
            "2019-20",
            "2020-21",
            "Overall_Growth_Percent",
            "Trend_Class"
        ]
    ]
)


print("\nClass Distribution:")

print(
    data[
        "Trend_Class"
    ]
    .value_counts()
)


# =============================================================================
# 19. SELECT FEATURES
# =============================================================================

FEATURE_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21",

    "Total_Allocation",
    "Average_Allocation",

    "Minimum_Allocation",
    "Maximum_Allocation",

    "Std_Allocation",

    "Growth_2018_2019",
    "Growth_2019_2020",

    "Overall_Growth",

    "Growth_Percent_2018_2019",
    "Growth_Percent_2019_2020",

    "Overall_Growth_Percent",

    "Trend_Slope",

    "Coefficient_of_Variation"
]


X = data[
    FEATURE_COLUMNS
].copy()


X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


X = X.fillna(0)


y = data[
    "Trend_Class"
].copy()


# =============================================================================
# 20. LABEL ENCODING
# =============================================================================

label_encoder = LabelEncoder()


y_encoded = (
    label_encoder
    .fit_transform(
        y
    )
)


print("\nClass Mapping:")

for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "->",
        class_name
    )


# =============================================================================
# 21. DEFINE MODELS
# =============================================================================

random_forest = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(

                n_estimators=300,

                max_depth=None,

                min_samples_split=2,

                random_state=42,

                class_weight="balanced"
            )
        )
    ]
)


gradient_boosting = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            GradientBoostingClassifier(

                n_estimators=150,

                learning_rate=0.05,

                max_depth=2,

                random_state=42
            )
        )
    ]
)


# =============================================================================
# 22. TRAIN MODELS ON COMPLETE DATA
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING MODELS")
print("=" * 90)


random_forest.fit(
    X,
    y_encoded
)


gradient_boosting.fit(
    X,
    y_encoded
)


# =============================================================================
# 23. TRAINING PREDICTIONS
# =============================================================================

rf_train_predictions = (
    random_forest
    .predict(
        X
    )
)


gb_train_predictions = (
    gradient_boosting
    .predict(
        X
    )
)


rf_train_accuracy = accuracy_score(

    y_encoded,

    rf_train_predictions
)


gb_train_accuracy = accuracy_score(

    y_encoded,

    gb_train_predictions
)


print(
    "\nRandom Forest Training Accuracy:",
    f"{rf_train_accuracy * 100:.2f}%"
)


print(
    "Gradient Boosting Training Accuracy:",
    f"{gb_train_accuracy * 100:.2f}%"
)


# =============================================================================
# 24. LEAVE-ONE-OUT CROSS VALIDATION
# =============================================================================

print("\n" + "=" * 90)
print("LEAVE-ONE-OUT CROSS VALIDATION")
print("=" * 90)


loo = LeaveOneOut()


rf_cv_actual = []
rf_cv_predicted = []


gb_cv_actual = []
gb_cv_predicted = []


for train_index, test_index in loo.split(X):

    X_train = X.iloc[
        train_index
    ]

    X_test = X.iloc[
        test_index
    ]

    y_train = y_encoded[
        train_index
    ]

    y_test = y_encoded[
        test_index
    ]


    # ---------------------------------------------------------
    # Check whether training fold has enough classes
    # ---------------------------------------------------------

    unique_classes = np.unique(
        y_train
    )


    if len(unique_classes) < 2:

        # Cannot train classifiers properly
        # Use majority class as fallback.

        prediction = unique_classes[0]

        rf_prediction = prediction
        gb_prediction = prediction


    else:

        # Random Forest
        rf_cv_model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    RandomForestClassifier(

                        n_estimators=300,

                        random_state=42,

                        class_weight="balanced"
                    )
                )
            ]
        )


        rf_cv_model.fit(
            X_train,
            y_train
        )


        rf_prediction = (
            rf_cv_model
            .predict(
                X_test
            )[0]
        )


        # Gradient Boosting
        gb_cv_model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    GradientBoostingClassifier(

                        n_estimators=150,

                        learning_rate=0.05,

                        max_depth=2,

                        random_state=42
                    )
                )
            ]
        )


        gb_cv_model.fit(
            X_train,
            y_train
        )


        gb_prediction = (
            gb_cv_model
            .predict(
                X_test
            )[0]
        )


    rf_cv_actual.append(
        y_test[0]
    )

    rf_cv_predicted.append(
        rf_prediction
    )


    gb_cv_actual.append(
        y_test[0]
    )

    gb_cv_predicted.append(
        gb_prediction
    )


# =============================================================================
# 25. CROSS VALIDATION ACCURACY
# =============================================================================

rf_cv_accuracy = accuracy_score(

    rf_cv_actual,

    rf_cv_predicted
)


gb_cv_accuracy = accuracy_score(

    gb_cv_actual,

    gb_cv_predicted
)


print(
    "\nRandom Forest LOOCV Accuracy:",
    f"{rf_cv_accuracy * 100:.2f}%"
)


print(
    "Gradient Boosting LOOCV Accuracy:",
    f"{gb_cv_accuracy * 100:.2f}%"
)


# =============================================================================
# 26. ACCURACY GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING ACCURACY GRAPH")
print("=" * 90)


model_names = [

    "Random Forest",
    "Gradient Boosting"
]


accuracy_values = [

    rf_cv_accuracy * 100,
    gb_cv_accuracy * 100
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(
    model_names,
    accuracy_values
)


plt.title(
    "Model Accuracy Comparison"
)


plt.xlabel(
    "Machine Learning Model"
)


plt.ylabel(
    "LOOCV Accuracy (%)"
)


plt.ylim(
    0,
    100
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    accuracy_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height() + 2,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    ACCURACY_GRAPH_PATH
)


# =============================================================================
# 27. CORRELATION HEATMAP
# =============================================================================

print("\n" + "=" * 90)
print("CREATING HEATMAP")
print("=" * 90)


heatmap_columns = [

    "2018-19",
    "2019-20",
    "2020-21",

    "Total_Allocation",

    "Average_Allocation",

    "Std_Allocation",

    "Overall_Growth",

    "Overall_Growth_Percent",

    "Trend_Slope"
]


correlation_matrix = (

    data[
        heatmap_columns
    ]
    .corr()
)


plt.figure(
    figsize=(13, 10)
)


plt.imshow(
    correlation_matrix,
    aspect="auto"
)


plt.colorbar(
    label="Correlation"
)


plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=45,

    ha="right"
)


plt.yticks(

    range(
        len(
            correlation_matrix.index
        )
    ),

    correlation_matrix.index
)


# Add correlation numbers
for i in range(
    len(
        correlation_matrix.index
    )
):

    for j in range(
        len(
            correlation_matrix.columns
        )
    ):

        value = (
            correlation_matrix
            .iloc[i, j]
        )

        if not np.isnan(value):

            plt.text(

                j,
                i,

                f"{value:.2f}",

                ha="center",
                va="center",

                fontsize=8
            )


plt.title(
    "Government Scheme Fund Allocation Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    HEATMAP_PATH
)


# =============================================================================
# 28. COMPARISON GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING YEAR-WISE COMPARISON GRAPH")
print("=" * 90)


x = np.arange(
    len(data)
)


width = 0.25


plt.figure(
    figsize=(15, 8)
)


plt.bar(

    x - width,

    data["2018-19"],

    width,

    label="2018-19"
)


plt.bar(

    x,

    data["2019-20"],

    width,

    label="2019-20"
)


plt.bar(

    x + width,

    data["2020-21"],

    width,

    label="2020-21"
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Fund Allocation"
)


plt.title(
    "Government Scheme Fund Allocation Comparison"
)


plt.xticks(

    x,

    data["Scheme"],

    rotation=45,

    ha="right"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    COMPARISON_GRAPH_PATH
)


# =============================================================================
# 29. CREATE RESULT CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING RESULT CSV")
print("=" * 90)


result_df = data[

    [

        "Scheme",

        "2018-19",
        "2019-20",
        "2020-21",

        "Total_Allocation",

        "Average_Allocation",

        "Minimum_Allocation",

        "Maximum_Allocation",

        "Std_Allocation",

        "Growth_2018_2019",

        "Growth_2019_2020",

        "Overall_Growth",

        "Growth_Percent_2018_2019",

        "Growth_Percent_2019_2020",

        "Overall_Growth_Percent",

        "Trend_Slope",

        "Coefficient_of_Variation",

        "Trend_Class"
    ]

].copy()


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nResult CSV saved:"
)


print(
    RESULT_CSV_PATH
)


print("\nResult Data:")

print(
    result_df
)


# =============================================================================
# 30. RESULT GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING RESULT GRAPH")
print("=" * 90)


plt.figure(
    figsize=(14, 7)
)


x = np.arange(
    len(result_df)
)


plt.bar(

    x,

    result_df[
        "Overall_Growth_Percent"
    ]
)


plt.axhline(
    y=0,
    linewidth=1
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Overall Growth (%)"
)


plt.title(
    "Government Scheme Overall Fund Growth"
)


plt.xticks(

    x,

    result_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


# Add values
for i, value in enumerate(

    result_df[
        "Overall_Growth_Percent"
    ]
):

    if value >= 0:

        vertical_alignment = "bottom"

    else:

        vertical_alignment = "top"


    plt.text(

        i,

        value,

        f"{value:.1f}%",

        ha="center",

        va=vertical_alignment,

        fontsize=8
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    RESULT_GRAPH_PATH
)


# =============================================================================
# 31. CREATE PREDICTIONS
# =============================================================================

print("\n" + "=" * 90)
print("CREATING MODEL PREDICTIONS")
print("=" * 90)


# We use LOOCV Random Forest predictions because
# they are more meaningful than predictions from a model
# evaluated on the same data used for training.


rf_cv_predicted = np.array(
    rf_cv_predicted
)


predicted_labels = (
    label_encoder
    .inverse_transform(
        rf_cv_predicted
    )
)


actual_labels = (
    label_encoder
    .inverse_transform(
        np.array(
            rf_cv_actual
        )
    )
)


prediction_df = pd.DataFrame(

    {

        "Scheme":
            data[
                "Scheme"
            ],

        "Actual_Trend":
            actual_labels,

        "Predicted_Trend":
            predicted_labels,

        "Correct_Prediction":
            (
                actual_labels
                ==
                predicted_labels
            )
    }
)


# Add encoded values for plotting
prediction_df[
    "Actual_Encoded"
] = np.array(
    rf_cv_actual
)


prediction_df[
    "Predicted_Encoded"
] = np.array(
    rf_cv_predicted
)


# =============================================================================
# 32. SAVE PREDICTION CSV
# =============================================================================

prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nPrediction CSV saved:"
)


print(
    PREDICTION_CSV_PATH
)


print("\nPrediction Data:")

print(
    prediction_df
)


# =============================================================================
# 33. PREDICTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PREDICTION GRAPH")
print("=" * 90)


x = np.arange(
    len(
        prediction_df
    )
)


width = 0.35


plt.figure(
    figsize=(14, 7)
)


plt.bar(

    x - width / 2,

    prediction_df[
        "Actual_Encoded"
    ],

    width,

    label="Actual"
)


plt.bar(

    x + width / 2,

    prediction_df[
        "Predicted_Encoded"
    ],

    width,

    label="Predicted"
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Trend Class"
)


plt.title(
    "Actual vs Predicted Government Scheme Funding Trend"
)


plt.xticks(

    x,

    prediction_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


class_positions = np.arange(
    len(
        label_encoder.classes_
    )
)


plt.yticks(

    class_positions,

    label_encoder.classes_
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    PREDICTION_GRAPH_PATH
)


# =============================================================================
# 34. CONFUSION MATRIX
# =============================================================================

print("\n" + "=" * 90)
print("RANDOM FOREST CONFUSION MATRIX")
print("=" * 90)


cm = confusion_matrix(

    rf_cv_actual,

    rf_cv_predicted,

    labels=np.arange(
        len(
            label_encoder.classes_
        )
    )
)


print(cm)


# =============================================================================
# 35. CLASSIFICATION REPORT
# =============================================================================

print("\n" + "=" * 90)
print("CLASSIFICATION REPORT")
print("=" * 90)


print(

    classification_report(

        rf_cv_actual,

        rf_cv_predicted,

        labels=np.arange(
            len(
                label_encoder.classes_
            )
        ),

        target_names=
            label_encoder.classes_,

        zero_division=0
    )
)


# =============================================================================
# 36. MODEL COMPARISON SUMMARY
# =============================================================================

comparison_df = pd.DataFrame(

    {

        "Model": [

            "Random Forest",

            "Gradient Boosting"
        ],

        "Training_Accuracy": [

            rf_train_accuracy,

            gb_train_accuracy
        ],

        "LOOCV_Accuracy": [

            rf_cv_accuracy,

            gb_cv_accuracy
        ]
    }
)


print("\n" + "=" * 90)
print("MODEL COMPARISON")
print("=" * 90)


print(
    comparison_df
)


# =============================================================================
# 37. VERIFY GENERATED FILES
# =============================================================================

print("\n" + "=" * 90)
print("VERIFYING GENERATED FILES")
print("=" * 90)


generated_files = [

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH
]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        size_kb = (

            os.path.getsize(
                file_path
            )

            / 1024
        )


        print(

            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )


    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(file_path)}"
        )


# =============================================================================
# 38. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("FINAL SUMMARY")
print("=" * 90)


print(
    "\nDataset Rows:",
    len(data)
)


print(
    "Number of Features:",
    len(FEATURE_COLUMNS)
)


print(
    "Trend Classes:",
    list(
        label_encoder.classes_
    )
)


print(
    "\nRandom Forest:"
)


print(
    "Training Accuracy:",
    f"{rf_train_accuracy * 100:.2f}%"
)


print(
    "LOOCV Accuracy:",
    f"{rf_cv_accuracy * 100:.2f}%"
)


print(
    "\nGradient Boosting:"
)


print(
    "Training Accuracy:",
    f"{gb_train_accuracy * 100:.2f}%"
)


print(
    "LOOCV Accuracy:",
    f"{gb_cv_accuracy * 100:.2f}%"
)


print("\nGenerated Files:")


for file_path in generated_files:

    print(
        "-",
        file_path
    )


print("\n" + "=" * 90)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 90)
```
