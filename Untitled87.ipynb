```python
import os
import json
import pickle
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yaml

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
from sklearn.model_selection import LeaveOneOut, StratifiedKFold, cross_val_score

warnings.filterwarnings("ignore")


# =============================================================================
# 2. RANDOM SEED
# =============================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


# =============================================================================
# 3. PATHS
# =============================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Scheme Fund Allocation Trend Analysis"
)

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_256_AU_3172_C.csv"
)

os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# =============================================================================
# 4. AIS OUTPUT PATHS
# =============================================================================

AIS_MODEL_PATH = os.path.join(
    BASE_DIR,
    "ais_model.pkl"
)

AIS_CONFIG_PATH = os.path.join(
    BASE_DIR,
    "ais_config.yaml"
)

AIS_METADATA_PATH = os.path.join(
    BASE_DIR,
    "ais_metadata.json"
)


AIS_ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_accuracy_graph.png"
)

AIS_HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "ais_heatmap.png"
)

AIS_COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_comparison_graph.png"
)

AIS_FEATURE_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_feature_selection_graph.png"
)

AIS_CLASS_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_class_distribution_graph.png"
)


AIS_RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "ais_result.csv"
)

AIS_RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_result_graph.png"
)


AIS_PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "ais_prediction.csv"
)

AIS_PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "ais_prediction_graph.png"
)


AIS_MODEL_COMPARISON_PATH = os.path.join(
    BASE_DIR,
    "ais_model_comparison.csv"
)

AIS_SELECTED_FEATURES_PATH = os.path.join(
    BASE_DIR,
    "ais_selected_features.csv"
)


print("=" * 90)
print("AIS GOVERNMENT SCHEME FUND ALLOCATION TREND ANALYSIS")
print("=" * 90)

print("\nDataset:")
print(CSV_PATH)

print("\nOutput Directory:")
print(BASE_DIR)


# =============================================================================
# 5. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


# =============================================================================
# 6. LOAD DATASET
# =============================================================================

print("\n" + "=" * 90)
print("LOADING DATASET")
print("=" * 90)


df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)


print("\nDataset loaded successfully.")

print(
    "\nDataset Shape:",
    df.shape
)


print("\nOriginal Columns:")

for i, col in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i}. {repr(col)}"
    )


print("\nFirst Rows:")

print(
    df.head()
)


# =============================================================================
# 7. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


df = df.dropna(
    axis=0,
    how="all"
)


df = df.dropna(
    axis=1,
    how="all"
)


df = df.reset_index(
    drop=True
)


print("\nCleaned Columns:")

for column in df.columns:

    print(
        "-",
        column
    )


# =============================================================================
# 8. NUMERIC CLEANING FUNCTION
# =============================================================================

def clean_numeric_series(series):

    cleaned = (
        series
        .astype(str)
        .str.strip()
        .str.replace(
            ",",
            "",
            regex=False
        )
        .str.replace(
            "₹",
            "",
            regex=False
        )
        .str.replace(
            "%",
            "",
            regex=False
        )
        .str.replace(
            "Rs.",
            "",
            regex=False
        )
        .str.replace(
            "Rs",
            "",
            regex=False
        )
    )


    cleaned = cleaned.replace(

        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],

        np.nan
    )


    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


# =============================================================================
# 9. NORMALIZE COLUMN NAME
# =============================================================================

def normalize_column_name(name):

    name = str(name).lower()

    name = (
        name
        .replace(
            "–",
            "-"
        )
        .replace(
            "—",
            "-"
        )
        .replace(
            "_",
            "-"
        )
        .replace(
            "/",
            "-"
        )
        .replace(
            " ",
            ""
        )
    )

    return name


# =============================================================================
# 10. IDENTIFY FINANCIAL YEAR COLUMNS
# =============================================================================

YEAR_PATTERNS = {

    "2018-19": [
        "2018-19",
        "201819"
    ],

    "2019-20": [
        "2019-20",
        "201920"
    ],

    "2020-21": [
        "2020-21",
        "202021"
    ]
}


detected_year_columns = {}


for column in df.columns:

    normalized = normalize_column_name(
        column
    )

    for year, patterns in YEAR_PATTERNS.items():

        for pattern in patterns:

            if pattern in normalized:

                detected_year_columns[
                    year
                ] = column

                break


print("\nDetected Financial-Year Columns:")


for year, column in detected_year_columns.items():

    print(
        year,
        "->",
        column
    )


# =============================================================================
# 11. FALLBACK NUMERIC COLUMN DETECTION
# =============================================================================

if len(detected_year_columns) < 3:

    print(
        "\nFinancial-year names not fully detected."
    )

    print(
        "Using automatic numeric-column detection..."
    )


    numeric_candidates = []


    for column in df.columns:

        converted = clean_numeric_series(
            df[column]
        )


        valid_ratio = (
            converted
            .notna()
            .mean()
        )


        if valid_ratio >= 0.50:

            numeric_candidates.append(
                column
            )


    print(
        "\nNumeric Candidates:",
        numeric_candidates
    )


    if len(numeric_candidates) < 3:

        raise ValueError(
            "Unable to find three numeric year columns."
        )


    selected = numeric_candidates[-3:]


    detected_year_columns = {

        "2018-19":
            selected[0],

        "2019-20":
            selected[1],

        "2020-21":
            selected[2]
    }


# =============================================================================
# 12. IDENTIFY SCHEME COLUMN
# =============================================================================

year_original_columns = list(
    detected_year_columns.values()
)


non_year_columns = [

    column

    for column in df.columns

    if column not in year_original_columns
]


if len(non_year_columns) > 0:

    scheme_column = (
        non_year_columns[0]
    )

else:

    scheme_column = None


# =============================================================================
# 13. CREATE CLEAN DATASET
# =============================================================================

data = pd.DataFrame()


if scheme_column is not None:

    data["Scheme"] = (
        df[scheme_column]
        .astype(str)
        .str.strip()
    )

else:

    data["Scheme"] = [

        f"Scheme_{i + 1}"

        for i in range(
            len(df)
        )
    ]


YEAR_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21"
]


for year in YEAR_COLUMNS:

    original_column = (
        detected_year_columns[
            year
        ]
    )


    data[year] = clean_numeric_series(
        df[original_column]
    )


# =============================================================================
# 14. REMOVE COMPLETELY INVALID ROWS
# =============================================================================

data = data.dropna(
    subset=YEAR_COLUMNS,
    how="all"
)


data = data.reset_index(
    drop=True
)


# =============================================================================
# 15. HANDLE MISSING VALUES
# =============================================================================

row_means = (
    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


for column in YEAR_COLUMNS:

    data[column] = (
        data[column]
        .fillna(
            row_means
        )
    )


for column in YEAR_COLUMNS:

    median_value = (
        data[column]
        .median()
    )


    data[column] = (
        data[column]
        .fillna(
            median_value
        )
    )


data = data.reset_index(
    drop=True
)


print("\nClean Dataset:")

print(
    data
)


print(
    "\nNumber of Schemes:",
    len(data)
)


# =============================================================================
# 16. FEATURE ENGINEERING
# =============================================================================

print("\n" + "=" * 90)
print("FEATURE ENGINEERING")
print("=" * 90)


EPSILON = 1e-9


# =============================================================================
# TOTAL
# =============================================================================

data[
    "Total_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .sum(axis=1)
)


# =============================================================================
# AVERAGE
# =============================================================================

data[
    "Average_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


# =============================================================================
# MINIMUM
# =============================================================================

data[
    "Minimum_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .min(axis=1)
)


# =============================================================================
# MAXIMUM
# =============================================================================

data[
    "Maximum_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .max(axis=1)
)


# =============================================================================
# STANDARD DEVIATION
# =============================================================================

data[
    "Std_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .std(axis=1)
)


# =============================================================================
# ABSOLUTE GROWTH
# =============================================================================

data[
    "Growth_2018_2019"
] = (
    data["2019-20"]
    -
    data["2018-19"]
)


data[
    "Growth_2019_2020"
] = (
    data["2020-21"]
    -
    data["2019-20"]
)


data[
    "Overall_Growth"
] = (
    data["2020-21"]
    -
    data["2018-19"]
)


# =============================================================================
# PERCENTAGE GROWTH
# =============================================================================

data[
    "Growth_Percent_2018_2019"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2019-20"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


data[
    "Growth_Percent_2019_2020"
] = np.where(

    np.abs(
        data["2019-20"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2019-20"]
        )
        /
        np.abs(
            data["2019-20"]
        )
    )
    * 100,

    0
)


data[
    "Overall_Growth_Percent"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


# =============================================================================
# 17. TREND SLOPE
# =============================================================================

def calculate_slope(row):

    x = np.array(
        [0, 1, 2],
        dtype=float
    )


    y = np.array(

        [
            row["2018-19"],
            row["2019-20"],
            row["2020-21"]
        ],

        dtype=float
    )


    slope = np.polyfit(
        x,
        y,
        1
    )[0]


    return float(
        slope
    )


data[
    "Trend_Slope"
] = data.apply(
    calculate_slope,
    axis=1
)


# =============================================================================
# 18. COEFFICIENT OF VARIATION
# =============================================================================

data[
    "Coefficient_of_Variation"
] = np.where(

    np.abs(
        data[
            "Average_Allocation"
        ]
    ) > EPSILON,

    (
        data[
            "Std_Allocation"
        ]
        /
        np.abs(
            data[
                "Average_Allocation"
            ]
        )
    ),

    0
)


# =============================================================================
# 19. TREND CLASSIFICATION
# =============================================================================

STABLE_THRESHOLD = 5.0


def classify_trend(row):

    y1 = row[
        "2018-19"
    ]

    y2 = row[
        "2019-20"
    ]

    y3 = row[
        "2020-21"
    ]


    growth = row[
        "Overall_Growth_Percent"
    ]


    if abs(growth) <= STABLE_THRESHOLD:

        return "Stable"


    elif y1 < y2 < y3:

        return "Increasing"


    elif y1 > y2 > y3:

        return "Decreasing"


    else:

        return "Fluctuating"


data[
    "Trend_Class"
] = data.apply(
    classify_trend,
    axis=1
)


print("\nTrend Classification:")


print(

    data[
        [
            "Scheme",
            "2018-19",
            "2019-20",
            "2020-21",
            "Overall_Growth_Percent",
            "Trend_Class"
        ]
    ]
)


print("\nClass Distribution:")

print(
    data[
        "Trend_Class"
    ]
    .value_counts()
)


# =============================================================================
# 20. FEATURE LIST
# =============================================================================

FEATURE_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21",

    "Total_Allocation",
    "Average_Allocation",

    "Minimum_Allocation",
    "Maximum_Allocation",

    "Std_Allocation",

    "Growth_2018_2019",
    "Growth_2019_2020",

    "Overall_Growth",

    "Growth_Percent_2018_2019",
    "Growth_Percent_2019_2020",

    "Overall_Growth_Percent",

    "Trend_Slope",

    "Coefficient_of_Variation"
]


X = data[
    FEATURE_COLUMNS
].copy()


X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


X = X.fillna(0)


y = data[
    "Trend_Class"
].copy()


# =============================================================================
# 21. LABEL ENCODING
# =============================================================================

label_encoder = LabelEncoder()


y_encoded = (
    label_encoder
    .fit_transform(
        y
    )
)


print("\nClass Mapping:")


for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "->",
        class_name
    )


# =============================================================================
# 22. AIS PARAMETERS
# =============================================================================

AIS_POPULATION_SIZE = 30

AIS_GENERATIONS = 25

AIS_CLONE_FACTOR = 5

AIS_MUTATION_RATE = 0.15

AIS_ELITE_COUNT = 5

AIS_RANDOM_NEW = 5


print("\n" + "=" * 90)
print("ARTIFICIAL IMMUNE SYSTEM CONFIGURATION")
print("=" * 90)


print(
    "\nPopulation Size:",
    AIS_POPULATION_SIZE
)


print(
    "Generations:",
    AIS_GENERATIONS
)


print(
    "Clone Factor:",
    AIS_CLONE_FACTOR
)


print(
    "Mutation Rate:",
    AIS_MUTATION_RATE
)


print(
    "Number of Features:",
    len(FEATURE_COLUMNS)
)


# =============================================================================
# 23. CREATE RANDOM ANTIBODY
# =============================================================================

def create_antibody(
    number_of_features
):

    antibody = np.random.randint(

        0,
        2,

        size=number_of_features
    )


    # Ensure at least one feature
    if antibody.sum() == 0:

        random_position = (
            np.random.randint(
                0,
                number_of_features
            )
        )

        antibody[
            random_position
        ] = 1


    return antibody


# =============================================================================
# 24. SELECT FEATURES FROM ANTIBODY
# =============================================================================

def get_selected_indices(
    antibody
):

    return np.where(
        antibody == 1
    )[0]


# =============================================================================
# 25. AIS FITNESS FUNCTION
# =============================================================================

def calculate_fitness(
    antibody,
    X_data,
    y_data
):

    selected_indices = (
        get_selected_indices(
            antibody
        )
    )


    if len(
        selected_indices
    ) == 0:

        return 0.0


    X_selected = (
        X_data.iloc[
            :,
            selected_indices
        ]
    )


    # ---------------------------------------------------------
    # Determine if stratified CV is possible
    # ---------------------------------------------------------

    unique_classes, counts = np.unique(
        y_data,
        return_counts=True
    )


    # ---------------------------------------------------------
    # If only one class exists
    # ---------------------------------------------------------

    if len(
        unique_classes
    ) < 2:

        return 0.0


    min_class_count = (
        counts.min()
    )


    # ---------------------------------------------------------
    # Model
    # ---------------------------------------------------------

    model = Pipeline(

        steps=[

            (
                "scaler",
                StandardScaler()
            ),

            (
                "classifier",
                RandomForestClassifier(

                    n_estimators=100,

                    random_state=RANDOM_STATE,

                    class_weight="balanced"
                )
            )
        ]
    )


    # ---------------------------------------------------------
    # CV score
    # ---------------------------------------------------------

    try:

        if min_class_count >= 2:

            number_of_splits = min(
                3,
                int(
                    min_class_count
                )
            )


            cv = StratifiedKFold(

                n_splits=number_of_splits,

                shuffle=True,

                random_state=RANDOM_STATE
            )


            scores = cross_val_score(

                model,

                X_selected,

                y_data,

                cv=cv,

                scoring="accuracy"
            )


            accuracy = (
                scores.mean()
            )


        else:

            # -------------------------------------------------
            # Small-data fallback:
            # resubstitution accuracy
            # -------------------------------------------------

            model.fit(
                X_selected,
                y_data
            )


            predictions = (
                model.predict(
                    X_selected
                )
            )


            accuracy = accuracy_score(

                y_data,

                predictions
            )


    except Exception:

        model.fit(
            X_selected,
            y_data
        )


        predictions = (
            model.predict(
                X_selected
            )
        )


        accuracy = accuracy_score(

            y_data,

            predictions
        )


    # ---------------------------------------------------------
    # Feature penalty
    # ---------------------------------------------------------

    feature_ratio = (

        len(
            selected_indices
        )

        /

        len(
            antibody
        )
    )


    # Prioritize accuracy but reward smaller feature sets.
    fitness = (

        0.95 * accuracy

        +

        0.05 * (
            1 - feature_ratio
        )
    )


    return float(
        fitness
    )


# =============================================================================
# 26. CLONE ANTIBODY
# =============================================================================

def clone_antibody(
    antibody,
    clone_factor
):

    return [

        antibody.copy()

        for _ in range(
            clone_factor
        )
    ]


# =============================================================================
# 27. MUTATE ANTIBODY
# =============================================================================

def mutate_antibody(
    antibody,
    mutation_rate
):

    mutated = (
        antibody.copy()
    )


    for i in range(
        len(mutated)
    ):

        if (
            np.random.random()
            <
            mutation_rate
        ):

            mutated[i] = (
                1 - mutated[i]
            )


    # Ensure at least one selected feature
    if mutated.sum() == 0:

        position = (
            np.random.randint(
                0,
                len(mutated)
            )
        )

        mutated[
            position
        ] = 1


    return mutated


# =============================================================================
# 28. INITIALIZE AIS POPULATION
# =============================================================================

print("\n" + "=" * 90)
print("INITIALIZING AIS POPULATION")
print("=" * 90)


population = [

    create_antibody(
        len(FEATURE_COLUMNS)
    )

    for _ in range(
        AIS_POPULATION_SIZE
    )
]


best_fitness_history = []

average_fitness_history = []

best_antibody = None

best_fitness = -1


# =============================================================================
# 29. RUN AIS / CLONAL SELECTION
# =============================================================================

print("\n" + "=" * 90)
print("RUNNING ARTIFICIAL IMMUNE SYSTEM")
print("=" * 90)


for generation in range(
    AIS_GENERATIONS
):

    # -------------------------------------------------------------------------
    # Evaluate population
    # -------------------------------------------------------------------------

    fitness_values = [

        calculate_fitness(
            antibody,
            X,
            y_encoded
        )

        for antibody in population
    ]


    # -------------------------------------------------------------------------
    # Sort by fitness
    # -------------------------------------------------------------------------

    sorted_indices = np.argsort(
        fitness_values
    )[::-1]


    population = [

        population[i]

        for i in sorted_indices
    ]


    fitness_values = [

        fitness_values[i]

        for i in sorted_indices
    ]


    generation_best = (
        fitness_values[0]
    )


    generation_average = (
        np.mean(
            fitness_values
        )
    )


    best_fitness_history.append(
        generation_best
    )


    average_fitness_history.append(
        generation_average
    )


    # -------------------------------------------------------------------------
    # Update global best
    # -------------------------------------------------------------------------

    if (
        generation_best
        >
        best_fitness
    ):

        best_fitness = (
            generation_best
        )

        best_antibody = (
            population[0]
            .copy()
        )


    print(

        f"Generation "
        f"{generation + 1:02d}/{AIS_GENERATIONS} | "
        f"Best Fitness = {generation_best:.4f} | "
        f"Average Fitness = {generation_average:.4f} | "
        f"Selected Features = {population[0].sum()}"
    )


    # -------------------------------------------------------------------------
    # Elite antibodies
    # -------------------------------------------------------------------------

    elites = [

        population[i].copy()

        for i in range(
            min(
                AIS_ELITE_COUNT,
                len(population)
            )
        )
    ]


    # -------------------------------------------------------------------------
    # Clone and mutate elites
    # -------------------------------------------------------------------------

    clone_population = []


    for rank, elite in enumerate(
        elites
    ):

        clones = clone_antibody(

            elite,

            AIS_CLONE_FACTOR
        )


        # Higher ranked antibody gets lower mutation
        adaptive_mutation = (

            AIS_MUTATION_RATE

            *
            (
                1
                +
                (
                    rank
                    /
                    max(
                        1,
                        len(elites)
                    )
                )
            )
        )


        for clone in clones:

            mutated_clone = mutate_antibody(

                clone,

                adaptive_mutation
            )


            clone_population.append(
                mutated_clone
            )


    # -------------------------------------------------------------------------
    # Add random antibodies for diversity
    # -------------------------------------------------------------------------

    random_antibodies = [

        create_antibody(
            len(FEATURE_COLUMNS)
        )

        for _ in range(
            AIS_RANDOM_NEW
        )
    ]


    # -------------------------------------------------------------------------
    # Candidate population
    # -------------------------------------------------------------------------

    candidates = (

        elites

        +

        clone_population

        +

        random_antibodies
    )


    # -------------------------------------------------------------------------
    # Evaluate candidates
    # -------------------------------------------------------------------------

    candidate_fitness = [

        calculate_fitness(
            antibody,
            X,
            y_encoded
        )

        for antibody in candidates
    ]


    candidate_order = np.argsort(
        candidate_fitness
    )[::-1]


    # -------------------------------------------------------------------------
    # Select next population
    # -------------------------------------------------------------------------

    population = [

        candidates[i]

        for i in candidate_order[
            :AIS_POPULATION_SIZE
        ]
    ]


# =============================================================================
# 30. AIS RESULTS
# =============================================================================

print("\n" + "=" * 90)
print("AIS OPTIMIZATION COMPLETE")
print("=" * 90)


selected_indices = (
    get_selected_indices(
        best_antibody
    )
)


selected_features = [

    FEATURE_COLUMNS[i]

    for i in selected_indices
]


print(
    "\nBest AIS Fitness:",
    round(
        best_fitness,
        4
    )
)


print(
    "\nSelected Features:",
    len(selected_features)
)


for feature in selected_features:

    print(
        "-",
        feature
    )


# =============================================================================
# 31. SAVE SELECTED FEATURES CSV
# =============================================================================

selected_features_df = pd.DataFrame(

    {

        "Feature":
            FEATURE_COLUMNS,

        "Selected":
            [
                int(value)
                for value
                in best_antibody
            ]
    }
)


selected_features_df.to_csv(

    AIS_SELECTED_FEATURES_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    AIS_SELECTED_FEATURES_PATH
)


# =============================================================================
# 32. SELECT OPTIMAL DATASET
# =============================================================================

X_ais = X[
    selected_features
].copy()


print(
    "\nAIS Optimized Feature Matrix Shape:",
    X_ais.shape
)


# =============================================================================
# 33. TRAIN FINAL AIS-RANDOM FOREST MODEL
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING AIS-OPTIMIZED RANDOM FOREST")
print("=" * 90)


ais_model = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(

                n_estimators=500,

                random_state=RANDOM_STATE,

                class_weight="balanced"
            )
        )
    ]
)


ais_model.fit(
    X_ais,
    y_encoded
)


ais_train_predictions = (
    ais_model.predict(
        X_ais
    )
)


ais_training_accuracy = (
    accuracy_score(

        y_encoded,

        ais_train_predictions
    )
)


print(
    "\nAIS Training Accuracy:",
    f"{ais_training_accuracy * 100:.2f}%"
)


# =============================================================================
# 34. LOOCV FOR AIS MODEL
# =============================================================================

print("\n" + "=" * 90)
print("AIS LEAVE-ONE-OUT CROSS VALIDATION")
print("=" * 90)


loo = LeaveOneOut()


ais_actual = []

ais_predicted = []


for train_index, test_index in loo.split(
    X_ais
):

    X_train = X_ais.iloc[
        train_index
    ]


    X_test = X_ais.iloc[
        test_index
    ]


    y_train = y_encoded[
        train_index
    ]


    y_test = y_encoded[
        test_index
    ]


    unique_train_classes = (
        np.unique(
            y_train
        )
    )


    # -------------------------------------------------------------------------
    # If only one class exists in training fold
    # -------------------------------------------------------------------------

    if len(
        unique_train_classes
    ) < 2:

        prediction = (
            unique_train_classes[0]
        )


    else:

        cv_model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    RandomForestClassifier(

                        n_estimators=500,

                        random_state=RANDOM_STATE,

                        class_weight="balanced"
                    )
                )
            ]
        )


        cv_model.fit(
            X_train,
            y_train
        )


        prediction = (
            cv_model
            .predict(
                X_test
            )[0]
        )


    ais_actual.append(
        y_test[0]
    )


    ais_predicted.append(
        prediction
    )


ais_actual = np.array(
    ais_actual
)


ais_predicted = np.array(
    ais_predicted
)


ais_cv_accuracy = accuracy_score(

    ais_actual,

    ais_predicted
)


print(
    "\nAIS LOOCV Accuracy:",
    f"{ais_cv_accuracy * 100:.2f}%"
)


# =============================================================================
# 35. BASELINE RANDOM FOREST FOR COMPARISON
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING BASELINE MODEL FOR COMPARISON")
print("=" * 90)


baseline_model = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(

                n_estimators=500,

                random_state=RANDOM_STATE,

                class_weight="balanced"
            )
        )
    ]
)


baseline_model.fit(
    X,
    y_encoded
)


baseline_train_predictions = (
    baseline_model.predict(
        X
    )
)


baseline_training_accuracy = (
    accuracy_score(

        y_encoded,

        baseline_train_predictions
    )
)


# =============================================================================
# 36. BASELINE LOOCV
# =============================================================================

baseline_actual = []

baseline_predicted = []


for train_index, test_index in LeaveOneOut().split(
    X
):

    X_train = X.iloc[
        train_index
    ]


    X_test = X.iloc[
        test_index
    ]


    y_train = y_encoded[
        train_index
    ]


    y_test = y_encoded[
        test_index
    ]


    unique_train_classes = (
        np.unique(
            y_train
        )
    )


    if len(
        unique_train_classes
    ) < 2:

        prediction = (
            unique_train_classes[0]
        )


    else:

        model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    RandomForestClassifier(

                        n_estimators=500,

                        random_state=RANDOM_STATE,

                        class_weight="balanced"
                    )
                )
            ]
        )


        model.fit(
            X_train,
            y_train
        )


        prediction = (
            model.predict(
                X_test
            )[0]
        )


    baseline_actual.append(
        y_test[0]
    )


    baseline_predicted.append(
        prediction
    )


baseline_cv_accuracy = (
    accuracy_score(

        baseline_actual,

        baseline_predicted
    )
)


print(
    "\nBaseline RF LOOCV Accuracy:",
    f"{baseline_cv_accuracy * 100:.2f}%"
)


print(
    "AIS-RF LOOCV Accuracy:",
    f"{ais_cv_accuracy * 100:.2f}%"
)


# =============================================================================
# 37. SAVE AIS MODEL
# =============================================================================

print("\n" + "=" * 90)
print("SAVING AIS MODEL")
print("=" * 90)


ais_model_package = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "optimization":
        "Artificial Immune System - Clonal Selection",

    "classifier":
        "RandomForestClassifier",

    "model":
        ais_model,

    "label_encoder":
        label_encoder,

    "all_features":
        FEATURE_COLUMNS,

    "selected_features":
        selected_features,

    "best_antibody":
        best_antibody.tolist(),

    "best_fitness":
        float(
            best_fitness
        ),

    "training_accuracy":
        float(
            ais_training_accuracy
        ),

    "loocv_accuracy":
        float(
            ais_cv_accuracy
        ),

    "classes":
        label_encoder.classes_.tolist(),

    "stable_threshold":
        STABLE_THRESHOLD
}


with open(
    AIS_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        ais_model_package,
        file
    )


print(
    "\nSaved:",
    AIS_MODEL_PATH
)


# =============================================================================
# 38. AIS ACCURACY GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS ACCURACY GRAPH")
print("=" * 90)


models = [

    "Baseline RF",
    "AIS + RF"
]


accuracy_values = [

    baseline_cv_accuracy * 100,
    ais_cv_accuracy * 100
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    models,

    accuracy_values
)


plt.title(
    "Baseline vs AIS Optimized Model Accuracy"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "LOOCV Accuracy (%)"
)


plt.ylim(
    0,
    100
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    accuracy_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height() + 2,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.tight_layout()


plt.savefig(

    AIS_ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_ACCURACY_GRAPH_PATH
)


# =============================================================================
# 39. AIS FEATURE SELECTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS FEATURE SELECTION GRAPH")
print("=" * 90)


feature_status = [

    "Selected"
    if value == 1
    else "Not Selected"

    for value in best_antibody
]


plt.figure(
    figsize=(15, 8)
)


x_positions = np.arange(
    len(FEATURE_COLUMNS)
)


plt.bar(

    x_positions,

    best_antibody
)


plt.xlabel(
    "Feature"
)


plt.ylabel(
    "AIS Selection"
)


plt.title(
    "AIS Feature Selection Results"
)


plt.xticks(

    x_positions,

    FEATURE_COLUMNS,

    rotation=60,

    ha="right"
)


plt.yticks(
    [0, 1],
    [
        "Not Selected",
        "Selected"
    ]
)


plt.tight_layout()


plt.savefig(

    AIS_FEATURE_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_FEATURE_GRAPH_PATH
)


# =============================================================================
# 40. AIS FITNESS / COMPARISON GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS COMPARISON GRAPH")
print("=" * 90)


generations = np.arange(
    1,
    AIS_GENERATIONS + 1
)


plt.figure(
    figsize=(11, 6)
)


plt.plot(

    generations,

    best_fitness_history,

    marker="o",

    label="Best Fitness"
)


plt.plot(

    generations,

    average_fitness_history,

    marker="s",

    label="Average Fitness"
)


plt.xlabel(
    "AIS Generation"
)


plt.ylabel(
    "Fitness"
)


plt.title(
    "AIS Optimization Progress"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    AIS_COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_COMPARISON_GRAPH_PATH
)


# =============================================================================
# 41. AIS CORRELATION HEATMAP
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS HEATMAP")
print("=" * 90)


heatmap_features = (
    selected_features.copy()
)


if len(
    heatmap_features
) < 2:

    heatmap_features = (
        FEATURE_COLUMNS[:5]
    )


correlation_matrix = (
    data[
        heatmap_features
    ]
    .corr()
)


plt.figure(
    figsize=(12, 10)
)


plt.imshow(
    correlation_matrix,
    aspect="auto"
)


plt.colorbar(
    label="Correlation"
)


plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=60,

    ha="right"
)


plt.yticks(

    range(
        len(
            correlation_matrix.index
        )
    ),

    correlation_matrix.index
)


for i in range(
    len(
        correlation_matrix.index
    )
):

    for j in range(
        len(
            correlation_matrix.columns
        )
    ):

        value = (
            correlation_matrix
            .iloc[i, j]
        )


        if not np.isnan(
            value
        ):

            plt.text(

                j,
                i,

                f"{value:.2f}",

                ha="center",
                va="center",

                fontsize=8
            )


plt.title(
    "AIS Selected Features Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(

    AIS_HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_HEATMAP_PATH
)


# =============================================================================
# 42. CLASS DISTRIBUTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS CLASS DISTRIBUTION GRAPH")
print("=" * 90)


class_counts = (
    data[
        "Trend_Class"
    ]
    .value_counts()
)


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    class_counts.index,

    class_counts.values
)


plt.title(
    "Government Scheme Trend Class Distribution"
)


plt.xlabel(
    "Trend Class"
)


plt.ylabel(
    "Number of Schemes"
)


for bar, value in zip(
    bars,
    class_counts.values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 0.05,

        str(
            value
        ),

        ha="center",

        fontweight="bold"
    )


plt.tight_layout()


plt.savefig(

    AIS_CLASS_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_CLASS_GRAPH_PATH
)


# =============================================================================
# 43. CREATE AIS RESULT CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS RESULT CSV")
print("=" * 90)


ais_result_df = data[

    [

        "Scheme",

        "2018-19",
        "2019-20",
        "2020-21",

        "Total_Allocation",

        "Average_Allocation",

        "Minimum_Allocation",

        "Maximum_Allocation",

        "Std_Allocation",

        "Growth_2018_2019",

        "Growth_2019_2020",

        "Overall_Growth",

        "Growth_Percent_2018_2019",

        "Growth_Percent_2019_2020",

        "Overall_Growth_Percent",

        "Trend_Slope",

        "Coefficient_of_Variation",

        "Trend_Class"
    ]

].copy()


ais_result_df[
    "AIS_Selected_Feature_Count"
] = len(
    selected_features
)


ais_result_df[
    "AIS_Fitness"
] = best_fitness


ais_result_df[
    "AIS_LOOCV_Accuracy"
] = ais_cv_accuracy


ais_result_df.to_csv(

    AIS_RESULT_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    AIS_RESULT_CSV_PATH
)


print("\nAIS Result:")

print(
    ais_result_df
)


# =============================================================================
# 44. AIS RESULT GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS RESULT GRAPH")
print("=" * 90)


plt.figure(
    figsize=(14, 7)
)


x_positions = np.arange(
    len(
        ais_result_df
    )
)


values = (
    ais_result_df[
        "Overall_Growth_Percent"
    ]
)


bars = plt.bar(

    x_positions,

    values
)


plt.axhline(
    y=0,
    linewidth=1
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Overall Growth (%)"
)


plt.title(
    "AIS Government Scheme Fund Growth Analysis"
)


plt.xticks(

    x_positions,

    ais_result_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


for i, value in enumerate(
    values
):

    if value >= 0:

        va = "bottom"

    else:

        va = "top"


    plt.text(

        i,

        value,

        f"{value:.1f}%",

        ha="center",

        va=va,

        fontsize=8
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    AIS_RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_RESULT_GRAPH_PATH
)


# =============================================================================
# 45. CREATE AIS PREDICTION CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS PREDICTION CSV")
print("=" * 90)


actual_labels = (
    label_encoder
    .inverse_transform(
        ais_actual
    )
)


predicted_labels = (
    label_encoder
    .inverse_transform(
        ais_predicted
    )
)


ais_prediction_df = pd.DataFrame(

    {

        "Scheme":
            data[
                "Scheme"
            ],

        "Actual_Trend":
            actual_labels,

        "AIS_Predicted_Trend":
            predicted_labels,

        "Correct_Prediction":
            (
                actual_labels
                ==
                predicted_labels
            ),

        "Actual_Encoded":
            ais_actual,

        "AIS_Predicted_Encoded":
            ais_predicted
    }
)


ais_prediction_df.to_csv(

    AIS_PREDICTION_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    AIS_PREDICTION_CSV_PATH
)


print("\nAIS Predictions:")

print(
    ais_prediction_df
)


# =============================================================================
# 46. AIS PREDICTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING AIS PREDICTION GRAPH")
print("=" * 90)


x_positions = np.arange(
    len(
        ais_prediction_df
    )
)


width = 0.35


plt.figure(
    figsize=(14, 7)
)


plt.bar(

    x_positions
    -
    width / 2,

    ais_prediction_df[
        "Actual_Encoded"
    ],

    width,

    label="Actual"
)


plt.bar(

    x_positions
    +
    width / 2,

    ais_prediction_df[
        "AIS_Predicted_Encoded"
    ],

    width,

    label="AIS Predicted"
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Trend Class"
)


plt.title(
    "AIS Actual vs Predicted Funding Trend"
)


plt.xticks(

    x_positions,

    ais_prediction_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


class_positions = np.arange(
    len(
        label_encoder.classes_
    )
)


plt.yticks(

    class_positions,

    label_encoder.classes_
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    AIS_PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    AIS_PREDICTION_GRAPH_PATH
)


# =============================================================================
# 47. SAVE MODEL COMPARISON CSV
# =============================================================================

comparison_df = pd.DataFrame(

    {

        "Model": [

            "Baseline Random Forest",

            "AIS Optimized Random Forest"
        ],

        "Number_of_Features": [

            len(
                FEATURE_COLUMNS
            ),

            len(
                selected_features
            )
        ],

        "Training_Accuracy": [

            baseline_training_accuracy,

            ais_training_accuracy
        ],

        "LOOCV_Accuracy": [

            baseline_cv_accuracy,

            ais_cv_accuracy
        ]
    }
)


comparison_df.to_csv(

    AIS_MODEL_COMPARISON_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    AIS_MODEL_COMPARISON_PATH
)


# =============================================================================
# 48. CONFUSION MATRIX
# =============================================================================

print("\n" + "=" * 90)
print("AIS CONFUSION MATRIX")
print("=" * 90)


cm = confusion_matrix(

    ais_actual,

    ais_predicted,

    labels=np.arange(
        len(
            label_encoder.classes_
        )
    )
)


print(
    cm
)


# =============================================================================
# 49. CLASSIFICATION REPORT
# =============================================================================

print("\n" + "=" * 90)
print("AIS CLASSIFICATION REPORT")
print("=" * 90)


print(

    classification_report(

        ais_actual,

        ais_predicted,

        labels=np.arange(
            len(
                label_encoder.classes_
            )
        ),

        target_names=
            label_encoder.classes_,

        zero_division=0
    )
)


# =============================================================================
# 50. CREATE AIS YAML CONFIG
# =============================================================================

ais_config = {

    "project": {

        "name":
            "AI-Based Government Scheme Fund Allocation Trend Analysis",

        "optimization":
            "Artificial Immune System",

        "ais_algorithm":
            "Clonal Selection",

        "classifier":
            "Random Forest"
    },


    "dataset": {

        "filename":
            "RS_Session_256_AU_3172_C.csv",

        "path":
            CSV_PATH,

        "rows":
            int(
                len(data)
            )
    },


    "ais_parameters": {

        "population_size":
            AIS_POPULATION_SIZE,

        "generations":
            AIS_GENERATIONS,

        "clone_factor":
            AIS_CLONE_FACTOR,

        "mutation_rate":
            AIS_MUTATION_RATE,

        "elite_count":
            AIS_ELITE_COUNT,

        "random_new_antibodies":
            AIS_RANDOM_NEW
    },


    "features": {

        "total_features":
            len(
                FEATURE_COLUMNS
            ),

        "selected_feature_count":
            len(
                selected_features
            ),

        "selected_features":
            selected_features
    },


    "performance": {

        "best_fitness":
            float(
                best_fitness
            ),

        "training_accuracy":
            float(
                ais_training_accuracy
            ),

        "loocv_accuracy":
            float(
                ais_cv_accuracy
            ),

        "baseline_loocv_accuracy":
            float(
                baseline_cv_accuracy
            )
    }
}


with open(

    AIS_CONFIG_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        ais_config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nSaved:",
    AIS_CONFIG_PATH
)


# =============================================================================
# 51. CREATE AIS JSON METADATA
# =============================================================================

ais_metadata = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "optimization_method":
        "Artificial Immune System",

    "ais_variant":
        "Clonal Selection Algorithm",

    "classifier":
        "Random Forest",

    "created_at":
        datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),

    "dataset": {

        "filename":
            os.path.basename(
                CSV_PATH
            ),

        "rows":
            int(
                len(data)
            ),

        "financial_years":
            YEAR_COLUMNS
    },


    "all_features":
        FEATURE_COLUMNS,


    "selected_features":
        selected_features,


    "selected_feature_count":
        int(
            len(
                selected_features
            )
        ),


    "best_antibody":
        best_antibody.tolist(),


    "best_fitness":
        float(
            best_fitness
        ),


    "training_accuracy":
        float(
            ais_training_accuracy
        ),


    "loocv_accuracy":
        float(
            ais_cv_accuracy
        ),


    "baseline_loocv_accuracy":
        float(
            baseline_cv_accuracy
        ),


    "classes":
        label_encoder.classes_.tolist(),


    "class_distribution": {

        str(key):
            int(value)

        for key, value in (
            data[
                "Trend_Class"
            ]
            .value_counts()
            .items()
        )
    },


    "output_prefix":
        "ais_"
}


with open(

    AIS_METADATA_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        ais_metadata,

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nSaved:",
    AIS_METADATA_PATH
)


# =============================================================================
# 52. VERIFY OUTPUT FILES
# =============================================================================

print("\n" + "=" * 90)
print("VERIFYING AIS OUTPUT FILES")
print("=" * 90)


generated_files = [

    AIS_MODEL_PATH,

    AIS_CONFIG_PATH,

    AIS_METADATA_PATH,

    AIS_ACCURACY_GRAPH_PATH,

    AIS_HEATMAP_PATH,

    AIS_COMPARISON_GRAPH_PATH,

    AIS_FEATURE_GRAPH_PATH,

    AIS_CLASS_GRAPH_PATH,

    AIS_RESULT_CSV_PATH,

    AIS_RESULT_GRAPH_PATH,

    AIS_PREDICTION_CSV_PATH,

    AIS_PREDICTION_GRAPH_PATH,

    AIS_MODEL_COMPARISON_PATH,

    AIS_SELECTED_FEATURES_PATH
]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        file_size = (
            os.path.getsize(
                file_path
            )
            /
            1024
        )


        print(

            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({file_size:.2f} KB)"
        )


    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(file_path)}"
        )


# =============================================================================
# 53. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("AIS FINAL SUMMARY")
print("=" * 90)


print(
    "\nDataset Rows:",
    len(data)
)


print(
    "Original Feature Count:",
    len(
        FEATURE_COLUMNS
    )
)


print(
    "AIS Selected Feature Count:",
    len(
        selected_features
    )
)


print(
    "\nSelected Features:"
)


for feature in selected_features:

    print(
        "-",
        feature
    )


print(
    "\nBest AIS Fitness:",
    f"{best_fitness:.4f}"
)


print(
    "Baseline Training Accuracy:",
    f"{baseline_training_accuracy * 100:.2f}%"
)


print(
    "AIS Training Accuracy:",
    f"{ais_training_accuracy * 100:.2f}%"
)


print(
    "Baseline LOOCV Accuracy:",
    f"{baseline_cv_accuracy * 100:.2f}%"
)


print(
    "AIS LOOCV Accuracy:",
    f"{ais_cv_accuracy * 100:.2f}%"
)


improvement = (

    (
        ais_cv_accuracy
        -
        baseline_cv_accuracy
    )

    * 100
)


print(
    "AIS Accuracy Difference:",
    f"{improvement:+.2f} percentage points"
)


print("\nGenerated AIS Files:")


for file_path in generated_files:

    print(
        "-",
        os.path.basename(
            file_path
        )
    )


print("\n" + "=" * 90)
print("AIS PROCESS COMPLETED SUCCESSFULLY")
print("=" * 90)
```
