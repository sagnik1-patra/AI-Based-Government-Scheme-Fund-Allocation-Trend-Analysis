```python
import os
import json
import pickle
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yaml

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
from sklearn.model_selection import (
    LeaveOneOut,
    StratifiedKFold,
    cross_val_score
)

warnings.filterwarnings("ignore")


# =============================================================================
# 2. RANDOM SEED
# =============================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


# =============================================================================
# 3. PATH CONFIGURATION
# =============================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Scheme Fund Allocation Trend Analysis"
)

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_256_AU_3172_C.csv"
)

os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# =============================================================================
# 4. PSO OUTPUT PATHS
# =============================================================================

PSO_MODEL_PATH = os.path.join(
    BASE_DIR,
    "pso_model.pkl"
)

PSO_CONFIG_PATH = os.path.join(
    BASE_DIR,
    "pso_config.yaml"
)

PSO_METADATA_PATH = os.path.join(
    BASE_DIR,
    "pso_metadata.json"
)


PSO_ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_accuracy_graph.png"
)

PSO_HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "pso_heatmap.png"
)

PSO_COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_comparison_graph.png"
)

PSO_FEATURE_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_feature_selection_graph.png"
)

PSO_CLASS_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_class_distribution_graph.png"
)


PSO_RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_result.csv"
)

PSO_RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_result_graph.png"
)


PSO_PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "pso_prediction.csv"
)

PSO_PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "pso_prediction_graph.png"
)


PSO_MODEL_COMPARISON_PATH = os.path.join(
    BASE_DIR,
    "pso_model_comparison.csv"
)

PSO_SELECTED_FEATURES_PATH = os.path.join(
    BASE_DIR,
    "pso_selected_features.csv"
)


# =============================================================================
# 5. DISPLAY PROJECT INFORMATION
# =============================================================================

print("=" * 90)
print("PSO GOVERNMENT SCHEME FUND ALLOCATION TREND ANALYSIS")
print("=" * 90)

print("\nDataset:")
print(CSV_PATH)

print("\nOutput Directory:")
print(BASE_DIR)


# =============================================================================
# 6. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n\n"
        "Please verify the CSV path."
    )


# =============================================================================
# 7. LOAD DATASET
# =============================================================================

print("\n" + "=" * 90)
print("LOADING DATASET")
print("=" * 90)


df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)


print("\nDataset loaded successfully.")

print(
    "\nDataset Shape:",
    df.shape
)


print("\nOriginal Columns:")

for i, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i}. {repr(column)}"
    )


print("\nFirst Rows:")

print(
    df.head()
)


# =============================================================================
# 8. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


df = df.dropna(
    axis=0,
    how="all"
)


df = df.dropna(
    axis=1,
    how="all"
)


df = df.reset_index(
    drop=True
)


print("\nCleaned Columns:")

for column in df.columns:

    print(
        "-",
        column
    )


# =============================================================================
# 9. NUMERIC CLEANING FUNCTION
# =============================================================================

def clean_numeric_series(series):

    cleaned = (
        series
        .astype(str)
        .str.strip()
        .str.replace(
            ",",
            "",
            regex=False
        )
        .str.replace(
            "₹",
            "",
            regex=False
        )
        .str.replace(
            "%",
            "",
            regex=False
        )
        .str.replace(
            "Rs.",
            "",
            regex=False
        )
        .str.replace(
            "Rs",
            "",
            regex=False
        )
    )


    cleaned = cleaned.replace(
        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],
        np.nan
    )


    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


# =============================================================================
# 10. NORMALIZE COLUMN NAME
# =============================================================================

def normalize_column_name(name):

    name = str(name).lower()

    name = (
        name
        .replace(
            "–",
            "-"
        )
        .replace(
            "—",
            "-"
        )
        .replace(
            "_",
            "-"
        )
        .replace(
            "/",
            "-"
        )
        .replace(
            " ",
            ""
        )
    )

    return name


# =============================================================================
# 11. IDENTIFY FINANCIAL YEAR COLUMNS
# =============================================================================

YEAR_PATTERNS = {

    "2018-19": [
        "2018-19",
        "201819"
    ],

    "2019-20": [
        "2019-20",
        "201920"
    ],

    "2020-21": [
        "2020-21",
        "202021"
    ]
}


detected_year_columns = {}


for column in df.columns:

    normalized = normalize_column_name(
        column
    )

    for year, patterns in YEAR_PATTERNS.items():

        for pattern in patterns:

            if pattern in normalized:

                detected_year_columns[
                    year
                ] = column

                break


print("\nDetected Financial-Year Columns:")


for year, column in detected_year_columns.items():

    print(
        year,
        "->",
        column
    )


# =============================================================================
# 12. FALLBACK NUMERIC COLUMN DETECTION
# =============================================================================

if len(detected_year_columns) < 3:

    print(
        "\nFinancial-year columns were not fully detected."
    )

    print(
        "Trying automatic numeric-column detection..."
    )


    numeric_candidates = []


    for column in df.columns:

        converted = clean_numeric_series(
            df[column]
        )


        valid_ratio = (
            converted
            .notna()
            .mean()
        )


        if valid_ratio >= 0.50:

            numeric_candidates.append(
                column
            )


    print(
        "\nNumeric Candidates:",
        numeric_candidates
    )


    if len(numeric_candidates) < 3:

        raise ValueError(
            "Unable to identify three numeric financial-year columns."
        )


    selected = numeric_candidates[-3:]


    detected_year_columns = {

        "2018-19":
            selected[0],

        "2019-20":
            selected[1],

        "2020-21":
            selected[2]
    }


# =============================================================================
# 13. IDENTIFY SCHEME COLUMN
# =============================================================================

year_original_columns = list(
    detected_year_columns.values()
)


non_year_columns = [

    column

    for column in df.columns

    if column not in year_original_columns
]


if len(non_year_columns) > 0:

    scheme_column = (
        non_year_columns[0]
    )

else:

    scheme_column = None


# =============================================================================
# 14. CREATE CLEAN DATAFRAME
# =============================================================================

data = pd.DataFrame()


if scheme_column is not None:

    data["Scheme"] = (
        df[scheme_column]
        .astype(str)
        .str.strip()
    )

else:

    data["Scheme"] = [

        f"Scheme_{i + 1}"

        for i in range(
            len(df)
        )
    ]


YEAR_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21"
]


for year in YEAR_COLUMNS:

    original_column = (
        detected_year_columns[
            year
        ]
    )


    data[year] = clean_numeric_series(
        df[original_column]
    )


# =============================================================================
# 15. REMOVE COMPLETELY EMPTY FINANCIAL ROWS
# =============================================================================

data = data.dropna(
    subset=YEAR_COLUMNS,
    how="all"
)


data = data.reset_index(
    drop=True
)


# =============================================================================
# 16. HANDLE MISSING VALUES
# =============================================================================

row_means = (
    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


for column in YEAR_COLUMNS:

    data[column] = (
        data[column]
        .fillna(
            row_means
        )
    )


for column in YEAR_COLUMNS:

    median_value = (
        data[column]
        .median()
    )


    data[column] = (
        data[column]
        .fillna(
            median_value
        )
    )


data = data.reset_index(
    drop=True
)


print("\nClean Dataset:")

print(
    data
)


print(
    "\nNumber of Schemes:",
    len(data)
)


# =============================================================================
# 17. FEATURE ENGINEERING
# =============================================================================

print("\n" + "=" * 90)
print("FEATURE ENGINEERING")
print("=" * 90)


EPSILON = 1e-9


# =============================================================================
# TOTAL ALLOCATION
# =============================================================================

data[
    "Total_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .sum(axis=1)
)


# =============================================================================
# AVERAGE ALLOCATION
# =============================================================================

data[
    "Average_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .mean(axis=1)
)


# =============================================================================
# MINIMUM ALLOCATION
# =============================================================================

data[
    "Minimum_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .min(axis=1)
)


# =============================================================================
# MAXIMUM ALLOCATION
# =============================================================================

data[
    "Maximum_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .max(axis=1)
)


# =============================================================================
# STANDARD DEVIATION
# =============================================================================

data[
    "Std_Allocation"
] = (
    data[
        YEAR_COLUMNS
    ]
    .std(axis=1)
)


# =============================================================================
# ABSOLUTE GROWTH
# =============================================================================

data[
    "Growth_2018_2019"
] = (
    data["2019-20"]
    -
    data["2018-19"]
)


data[
    "Growth_2019_2020"
] = (
    data["2020-21"]
    -
    data["2019-20"]
)


data[
    "Overall_Growth"
] = (
    data["2020-21"]
    -
    data["2018-19"]
)


# =============================================================================
# PERCENTAGE GROWTH
# =============================================================================

data[
    "Growth_Percent_2018_2019"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2019-20"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


data[
    "Growth_Percent_2019_2020"
] = np.where(

    np.abs(
        data["2019-20"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2019-20"]
        )
        /
        np.abs(
            data["2019-20"]
        )
    )
    * 100,

    0
)


data[
    "Overall_Growth_Percent"
] = np.where(

    np.abs(
        data["2018-19"]
    ) > EPSILON,

    (
        (
            data["2020-21"]
            -
            data["2018-19"]
        )
        /
        np.abs(
            data["2018-19"]
        )
    )
    * 100,

    0
)


# =============================================================================
# 18. TREND SLOPE
# =============================================================================

def calculate_slope(row):

    x = np.array(
        [0, 1, 2],
        dtype=float
    )


    y_values = np.array(
        [
            row["2018-19"],
            row["2019-20"],
            row["2020-21"]
        ],
        dtype=float
    )


    slope = np.polyfit(
        x,
        y_values,
        1
    )[0]


    return float(
        slope
    )


data[
    "Trend_Slope"
] = data.apply(
    calculate_slope,
    axis=1
)


# =============================================================================
# 19. COEFFICIENT OF VARIATION
# =============================================================================

data[
    "Coefficient_of_Variation"
] = np.where(

    np.abs(
        data[
            "Average_Allocation"
        ]
    ) > EPSILON,

    (
        data[
            "Std_Allocation"
        ]
        /
        np.abs(
            data[
                "Average_Allocation"
            ]
        )
    ),

    0
)


# =============================================================================
# 20. CREATE TREND CLASSIFICATION
# =============================================================================

STABLE_THRESHOLD = 5.0


def classify_trend(row):

    year1 = row[
        "2018-19"
    ]

    year2 = row[
        "2019-20"
    ]

    year3 = row[
        "2020-21"
    ]


    overall_growth = row[
        "Overall_Growth_Percent"
    ]


    if abs(
        overall_growth
    ) <= STABLE_THRESHOLD:

        return "Stable"


    elif (
        year1
        <
        year2
        <
        year3
    ):

        return "Increasing"


    elif (
        year1
        >
        year2
        >
        year3
    ):

        return "Decreasing"


    else:

        return "Fluctuating"


data[
    "Trend_Class"
] = data.apply(
    classify_trend,
    axis=1
)


print("\nTrend Classification:")


print(

    data[
        [
            "Scheme",
            "2018-19",
            "2019-20",
            "2020-21",
            "Overall_Growth_Percent",
            "Trend_Class"
        ]
    ]
)


print("\nClass Distribution:")


print(

    data[
        "Trend_Class"
    ]
    .value_counts()
)


# =============================================================================
# 21. DEFINE FEATURES
# =============================================================================

FEATURE_COLUMNS = [

    "2018-19",
    "2019-20",
    "2020-21",

    "Total_Allocation",

    "Average_Allocation",

    "Minimum_Allocation",

    "Maximum_Allocation",

    "Std_Allocation",

    "Growth_2018_2019",

    "Growth_2019_2020",

    "Overall_Growth",

    "Growth_Percent_2018_2019",

    "Growth_Percent_2019_2020",

    "Overall_Growth_Percent",

    "Trend_Slope",

    "Coefficient_of_Variation"
]


X = data[
    FEATURE_COLUMNS
].copy()


X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


X = X.fillna(0)


y = data[
    "Trend_Class"
].copy()


print(
    "\nTotal Features:",
    len(
        FEATURE_COLUMNS
    )
)


# =============================================================================
# 22. LABEL ENCODING
# =============================================================================

label_encoder = LabelEncoder()


y_encoded = (
    label_encoder
    .fit_transform(
        y
    )
)


print("\nClass Mapping:")


for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "->",
        class_name
    )


# =============================================================================
# 23. PSO PARAMETERS
# =============================================================================

PSO_NUM_PARTICLES = 30

PSO_ITERATIONS = 30

PSO_INERTIA_WEIGHT = 0.729

PSO_COGNITIVE_COEFFICIENT = 1.49445

PSO_SOCIAL_COEFFICIENT = 1.49445


print("\n" + "=" * 90)
print("PARTICLE SWARM OPTIMIZATION CONFIGURATION")
print("=" * 90)


print(
    "\nNumber of Particles:",
    PSO_NUM_PARTICLES
)


print(
    "Iterations:",
    PSO_ITERATIONS
)


print(
    "Inertia Weight:",
    PSO_INERTIA_WEIGHT
)


print(
    "Cognitive Coefficient:",
    PSO_COGNITIVE_COEFFICIENT
)


print(
    "Social Coefficient:",
    PSO_SOCIAL_COEFFICIENT
)


print(
    "Number of Features:",
    len(
        FEATURE_COLUMNS
    )
)


# =============================================================================
# 24. SIGMOID FUNCTION
# =============================================================================

def sigmoid(x):

    # Prevent overflow
    x = np.clip(
        x,
        -20,
        20
    )

    return (
        1.0
        /
        (
            1.0
            +
            np.exp(
                -x
            )
        )
    )


# =============================================================================
# 25. ENSURE AT LEAST ONE FEATURE
# =============================================================================

def ensure_valid_particle(
    position
):

    position = (
        position.copy()
    )


    if position.sum() == 0:

        random_feature = (
            np.random.randint(
                0,
                len(position)
            )
        )

        position[
            random_feature
        ] = 1


    return position


# =============================================================================
# 26. PSO FITNESS FUNCTION
# =============================================================================

def calculate_pso_fitness(
    particle_position,
    X_data,
    y_data
):

    particle_position = (
        ensure_valid_particle(
            particle_position
        )
    )


    selected_indices = np.where(
        particle_position == 1
    )[0]


    if len(
        selected_indices
    ) == 0:

        return 0.0


    X_selected = (
        X_data.iloc[
            :,
            selected_indices
        ]
    )


    unique_classes, counts = np.unique(
        y_data,
        return_counts=True
    )


    if len(
        unique_classes
    ) < 2:

        return 0.0


    model = Pipeline(

        steps=[

            (
                "scaler",
                StandardScaler()
            ),

            (
                "classifier",
                RandomForestClassifier(

                    n_estimators=100,

                    random_state=RANDOM_STATE,

                    class_weight="balanced"
                )
            )
        ]
    )


    min_class_count = (
        counts.min()
    )


    try:

        # ---------------------------------------------------------------------
        # Use stratified CV when enough examples exist
        # ---------------------------------------------------------------------

        if min_class_count >= 2:

            n_splits = min(
                3,
                int(
                    min_class_count
                )
            )


            cv = StratifiedKFold(

                n_splits=n_splits,

                shuffle=True,

                random_state=RANDOM_STATE
            )


            scores = cross_val_score(

                model,

                X_selected,

                y_data,

                cv=cv,

                scoring="accuracy"
            )


            model_accuracy = (
                scores.mean()
            )


        else:

            # -----------------------------------------------------------------
            # Small-data fallback
            # -----------------------------------------------------------------

            model.fit(
                X_selected,
                y_data
            )


            predictions = (
                model.predict(
                    X_selected
                )
            )


            model_accuracy = accuracy_score(

                y_data,

                predictions
            )


    except Exception:

        model.fit(
            X_selected,
            y_data
        )


        predictions = (
            model.predict(
                X_selected
            )
        )


        model_accuracy = accuracy_score(

            y_data,

            predictions
        )


    # -------------------------------------------------------------------------
    # Feature selection penalty
    # -------------------------------------------------------------------------

    selected_ratio = (

        len(
            selected_indices
        )

        /

        len(
            particle_position
        )
    )


    # 95% importance to predictive performance
    # 5% importance to selecting fewer features

    fitness = (

        0.95
        *
        model_accuracy

        +

        0.05
        *
        (
            1
            -
            selected_ratio
        )
    )


    return float(
        fitness
    )


# =============================================================================
# 27. INITIALIZE PARTICLE SWARM
# =============================================================================

print("\n" + "=" * 90)
print("INITIALIZING PSO PARTICLES")
print("=" * 90)


NUM_FEATURES = len(
    FEATURE_COLUMNS
)


# Binary particle positions
positions = np.random.randint(

    0,
    2,

    size=(
        PSO_NUM_PARTICLES,
        NUM_FEATURES
    )
)


# Ensure every particle selects at least one feature
for i in range(
    PSO_NUM_PARTICLES
):

    positions[i] = ensure_valid_particle(
        positions[i]
    )


# Continuous velocities
velocities = np.random.uniform(

    low=-1.0,

    high=1.0,

    size=(
        PSO_NUM_PARTICLES,
        NUM_FEATURES
    )
)


# Personal best
personal_best_positions = (
    positions.copy()
)


personal_best_fitness = np.full(

    PSO_NUM_PARTICLES,

    -np.inf
)


# Global best
global_best_position = None

global_best_fitness = -np.inf


# History
global_best_history = []

average_fitness_history = []

selected_feature_history = []


# =============================================================================
# 28. EVALUATE INITIAL PARTICLES
# =============================================================================

print("\nEvaluating initial swarm...")


for particle_index in range(
    PSO_NUM_PARTICLES
):

    fitness = calculate_pso_fitness(

        positions[
            particle_index
        ],

        X,

        y_encoded
    )


    personal_best_fitness[
        particle_index
    ] = fitness


    if (
        fitness
        >
        global_best_fitness
    ):

        global_best_fitness = (
            fitness
        )


        global_best_position = (
            positions[
                particle_index
            ]
            .copy()
        )


print(
    "\nInitial Global Best Fitness:",
    round(
        global_best_fitness,
        4
    )
)


# =============================================================================
# 29. RUN BINARY PARTICLE SWARM OPTIMIZATION
# =============================================================================

print("\n" + "=" * 90)
print("RUNNING PARTICLE SWARM OPTIMIZATION")
print("=" * 90)


for iteration in range(
    PSO_ITERATIONS
):

    iteration_fitness = []


    # =========================================================================
    # UPDATE EACH PARTICLE
    # =========================================================================

    for particle_index in range(
        PSO_NUM_PARTICLES
    ):

        # ---------------------------------------------------------------------
        # Random coefficients
        # ---------------------------------------------------------------------

        r1 = np.random.random(
            NUM_FEATURES
        )


        r2 = np.random.random(
            NUM_FEATURES
        )


        # ---------------------------------------------------------------------
        # Cognitive component
        # ---------------------------------------------------------------------

        cognitive_component = (

            PSO_COGNITIVE_COEFFICIENT

            *
            r1

            *
            (
                personal_best_positions[
                    particle_index
                ]
                -
                positions[
                    particle_index
                ]
            )
        )


        # ---------------------------------------------------------------------
        # Social component
        # ---------------------------------------------------------------------

        social_component = (

            PSO_SOCIAL_COEFFICIENT

            *
            r2

            *
            (
                global_best_position
                -
                positions[
                    particle_index
                ]
            )
        )


        # ---------------------------------------------------------------------
        # Velocity update
        # ---------------------------------------------------------------------

        velocities[
            particle_index
        ] = (

            PSO_INERTIA_WEIGHT
            *
            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        # Limit velocity
        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -6,

            6
        )


        # ---------------------------------------------------------------------
        # Convert velocity into probability
        # ---------------------------------------------------------------------

        probabilities = sigmoid(

            velocities[
                particle_index
            ]
        )


        # ---------------------------------------------------------------------
        # Binary position update
        # ---------------------------------------------------------------------

        random_values = np.random.random(
            NUM_FEATURES
        )


        new_position = (

            random_values
            <
            probabilities

        ).astype(
            int
        )


        new_position = ensure_valid_particle(
            new_position
        )


        positions[
            particle_index
        ] = new_position


        # ---------------------------------------------------------------------
        # Calculate fitness
        # ---------------------------------------------------------------------

        fitness = calculate_pso_fitness(

            positions[
                particle_index
            ],

            X,

            y_encoded
        )


        iteration_fitness.append(
            fitness
        )


        # ---------------------------------------------------------------------
        # Update personal best
        # ---------------------------------------------------------------------

        if (
            fitness
            >
            personal_best_fitness[
                particle_index
            ]
        ):

            personal_best_fitness[
                particle_index
            ] = fitness


            personal_best_positions[
                particle_index
            ] = (

                positions[
                    particle_index
                ]
                .copy()
            )


        # ---------------------------------------------------------------------
        # Update global best
        # ---------------------------------------------------------------------

        if (
            fitness
            >
            global_best_fitness
        ):

            global_best_fitness = (
                fitness
            )


            global_best_position = (

                positions[
                    particle_index
                ]
                .copy()
            )


    # =========================================================================
    # STORE ITERATION HISTORY
    # =========================================================================

    global_best_history.append(
        global_best_fitness
    )


    average_fitness_history.append(

        np.mean(
            iteration_fitness
        )
    )


    selected_feature_history.append(

        int(
            global_best_position.sum()
        )
    )


    print(

        f"Iteration "
        f"{iteration + 1:02d}/{PSO_ITERATIONS} | "
        f"Best Fitness = "
        f"{global_best_fitness:.4f} | "
        f"Average Fitness = "
        f"{np.mean(iteration_fitness):.4f} | "
        f"Selected Features = "
        f"{int(global_best_position.sum())}"
    )


# =============================================================================
# 30. PSO OPTIMIZATION RESULTS
# =============================================================================

print("\n" + "=" * 90)
print("PSO OPTIMIZATION COMPLETED")
print("=" * 90)


best_particle = (
    global_best_position.copy()
)


selected_indices = np.where(
    best_particle == 1
)[0]


selected_features = [

    FEATURE_COLUMNS[i]

    for i in selected_indices
]


print(
    "\nBest PSO Fitness:",
    round(
        global_best_fitness,
        4
    )
)


print(
    "\nSelected Feature Count:",
    len(
        selected_features
    )
)


print("\nPSO Selected Features:")


for feature in selected_features:

    print(
        "-",
        feature
    )


# =============================================================================
# 31. SAVE SELECTED FEATURES CSV
# =============================================================================

pso_selected_features_df = pd.DataFrame(

    {

        "Feature":
            FEATURE_COLUMNS,

        "PSO_Selected":
            [
                int(value)
                for value
                in best_particle
            ]
    }
)


pso_selected_features_df.to_csv(

    PSO_SELECTED_FEATURES_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    PSO_SELECTED_FEATURES_PATH
)


# =============================================================================
# 32. CREATE PSO-OPTIMIZED FEATURE MATRIX
# =============================================================================

X_pso = X[
    selected_features
].copy()


print(
    "\nOriginal Feature Matrix:",
    X.shape
)


print(
    "PSO Feature Matrix:",
    X_pso.shape
)


# =============================================================================
# 33. TRAIN FINAL PSO RANDOM FOREST MODEL
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING PSO-OPTIMIZED RANDOM FOREST")
print("=" * 90)


pso_model = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(

                n_estimators=500,

                random_state=RANDOM_STATE,

                class_weight="balanced"
            )
        )
    ]
)


pso_model.fit(
    X_pso,
    y_encoded
)


# =============================================================================
# 34. PSO TRAINING PREDICTIONS
# =============================================================================

pso_training_predictions = (
    pso_model.predict(
        X_pso
    )
)


pso_training_accuracy = accuracy_score(

    y_encoded,

    pso_training_predictions
)


print(
    "\nPSO Training Accuracy:",
    f"{pso_training_accuracy * 100:.2f}%"
)


# =============================================================================
# 35. PSO LEAVE-ONE-OUT CROSS VALIDATION
# =============================================================================

print("\n" + "=" * 90)
print("PSO LEAVE-ONE-OUT CROSS VALIDATION")
print("=" * 90)


loo = LeaveOneOut()


pso_actual = []

pso_predicted = []


for train_index, test_index in loo.split(
    X_pso
):

    X_train = X_pso.iloc[
        train_index
    ]


    X_test = X_pso.iloc[
        test_index
    ]


    y_train = y_encoded[
        train_index
    ]


    y_test = y_encoded[
        test_index
    ]


    unique_train_classes = np.unique(
        y_train
    )


    # -------------------------------------------------------------------------
    # Handle folds with only one class
    # -------------------------------------------------------------------------

    if len(
        unique_train_classes
    ) < 2:

        prediction = (
            unique_train_classes[0]
        )


    else:

        cv_model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    RandomForestClassifier(

                        n_estimators=500,

                        random_state=RANDOM_STATE,

                        class_weight="balanced"
                    )
                )
            ]
        )


        cv_model.fit(
            X_train,
            y_train
        )


        prediction = (
            cv_model
            .predict(
                X_test
            )[0]
        )


    pso_actual.append(
        y_test[0]
    )


    pso_predicted.append(
        prediction
    )


pso_actual = np.array(
    pso_actual
)


pso_predicted = np.array(
    pso_predicted
)


pso_cv_accuracy = accuracy_score(

    pso_actual,

    pso_predicted
)


print(
    "\nPSO LOOCV Accuracy:",
    f"{pso_cv_accuracy * 100:.2f}%"
)


# =============================================================================
# 36. TRAIN BASELINE RANDOM FOREST
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING BASELINE RANDOM FOREST")
print("=" * 90)


baseline_model = Pipeline(

    steps=[

        (
            "scaler",
            StandardScaler()
        ),

        (
            "classifier",
            RandomForestClassifier(

                n_estimators=500,

                random_state=RANDOM_STATE,

                class_weight="balanced"
            )
        )
    ]
)


baseline_model.fit(
    X,
    y_encoded
)


baseline_training_predictions = (
    baseline_model.predict(
        X
    )
)


baseline_training_accuracy = accuracy_score(

    y_encoded,

    baseline_training_predictions
)


print(
    "\nBaseline Training Accuracy:",
    f"{baseline_training_accuracy * 100:.2f}%"
)


# =============================================================================
# 37. BASELINE LOOCV
# =============================================================================

baseline_actual = []

baseline_predicted = []


for train_index, test_index in LeaveOneOut().split(
    X
):

    X_train = X.iloc[
        train_index
    ]


    X_test = X.iloc[
        test_index
    ]


    y_train = y_encoded[
        train_index
    ]


    y_test = y_encoded[
        test_index
    ]


    unique_train_classes = np.unique(
        y_train
    )


    if len(
        unique_train_classes
    ) < 2:

        prediction = (
            unique_train_classes[0]
        )


    else:

        baseline_cv_model = Pipeline(

            steps=[

                (
                    "scaler",
                    StandardScaler()
                ),

                (
                    "classifier",
                    RandomForestClassifier(

                        n_estimators=500,

                        random_state=RANDOM_STATE,

                        class_weight="balanced"
                    )
                )
            ]
        )


        baseline_cv_model.fit(
            X_train,
            y_train
        )


        prediction = (
            baseline_cv_model
            .predict(
                X_test
            )[0]
        )


    baseline_actual.append(
        y_test[0]
    )


    baseline_predicted.append(
        prediction
    )


baseline_actual = np.array(
    baseline_actual
)


baseline_predicted = np.array(
    baseline_predicted
)


baseline_cv_accuracy = accuracy_score(

    baseline_actual,

    baseline_predicted
)


print(
    "\nBaseline LOOCV Accuracy:",
    f"{baseline_cv_accuracy * 100:.2f}%"
)


print(
    "PSO LOOCV Accuracy:",
    f"{pso_cv_accuracy * 100:.2f}%"
)


# =============================================================================
# 38. SAVE PSO MODEL
# =============================================================================

print("\n" + "=" * 90)
print("SAVING PSO MODEL")
print("=" * 90)


pso_model_package = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "optimization":
        "Binary Particle Swarm Optimization",

    "classifier":
        "RandomForestClassifier",

    "model":
        pso_model,

    "label_encoder":
        label_encoder,

    "all_features":
        FEATURE_COLUMNS,

    "selected_features":
        selected_features,

    "best_particle":
        best_particle.tolist(),

    "best_fitness":
        float(
            global_best_fitness
        ),

    "training_accuracy":
        float(
            pso_training_accuracy
        ),

    "loocv_accuracy":
        float(
            pso_cv_accuracy
        ),

    "baseline_loocv_accuracy":
        float(
            baseline_cv_accuracy
        ),

    "classes":
        label_encoder.classes_.tolist(),

    "stable_threshold":
        STABLE_THRESHOLD
}


with open(
    PSO_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        pso_model_package,
        file
    )


print(
    "\nSaved:",
    PSO_MODEL_PATH
)


# =============================================================================
# 39. CREATE PSO ACCURACY GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO ACCURACY GRAPH")
print("=" * 90)


model_names = [

    "Baseline RF",
    "PSO + RF"
]


accuracy_values = [

    baseline_cv_accuracy * 100,
    pso_cv_accuracy * 100
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    model_names,

    accuracy_values
)


plt.title(
    "Baseline vs PSO Optimized Model Accuracy"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "LOOCV Accuracy (%)"
)


plt.ylim(
    0,
    100
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    accuracy_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height() + 2,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.tight_layout()


plt.savefig(

    PSO_ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_ACCURACY_GRAPH_PATH
)


# =============================================================================
# 40. PSO FEATURE SELECTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO FEATURE SELECTION GRAPH")
print("=" * 90)


x_positions = np.arange(
    len(
        FEATURE_COLUMNS
    )
)


plt.figure(
    figsize=(15, 8)
)


plt.bar(

    x_positions,

    best_particle
)


plt.xlabel(
    "Feature"
)


plt.ylabel(
    "PSO Selection"
)


plt.title(
    "PSO Feature Selection Results"
)


plt.xticks(

    x_positions,

    FEATURE_COLUMNS,

    rotation=60,

    ha="right"
)


plt.yticks(

    [0, 1],

    [
        "Not Selected",
        "Selected"
    ]
)


plt.tight_layout()


plt.savefig(

    PSO_FEATURE_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_FEATURE_GRAPH_PATH
)


# =============================================================================
# 41. PSO OPTIMIZATION / COMPARISON GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO COMPARISON GRAPH")
print("=" * 90)


iterations = np.arange(
    1,
    PSO_ITERATIONS + 1
)


plt.figure(
    figsize=(11, 6)
)


plt.plot(

    iterations,

    global_best_history,

    marker="o",

    label="Global Best Fitness"
)


plt.plot(

    iterations,

    average_fitness_history,

    marker="s",

    label="Average Swarm Fitness"
)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Fitness"
)


plt.title(
    "Particle Swarm Optimization Progress"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_COMPARISON_GRAPH_PATH
)


# =============================================================================
# 42. PSO HEATMAP
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO HEATMAP")
print("=" * 90)


heatmap_features = (
    selected_features.copy()
)


# Ensure meaningful heatmap if PSO selected only one feature
if len(
    heatmap_features
) < 2:

    heatmap_features = (
        FEATURE_COLUMNS[:5]
    )


correlation_matrix = (

    data[
        heatmap_features
    ]
    .corr()
)


plt.figure(
    figsize=(12, 10)
)


plt.imshow(
    correlation_matrix,
    aspect="auto"
)


plt.colorbar(
    label="Correlation"
)


plt.xticks(

    range(
        len(
            correlation_matrix.columns
        )
    ),

    correlation_matrix.columns,

    rotation=60,

    ha="right"
)


plt.yticks(

    range(
        len(
            correlation_matrix.index
        )
    ),

    correlation_matrix.index
)


for i in range(
    len(
        correlation_matrix.index
    )
):

    for j in range(
        len(
            correlation_matrix.columns
        )
    ):

        value = (
            correlation_matrix
            .iloc[i, j]
        )


        if not np.isnan(
            value
        ):

            plt.text(

                j,
                i,

                f"{value:.2f}",

                ha="center",

                va="center",

                fontsize=8
            )


plt.title(
    "PSO Selected Features Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(

    PSO_HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_HEATMAP_PATH
)


# =============================================================================
# 43. PSO CLASS DISTRIBUTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO CLASS DISTRIBUTION GRAPH")
print("=" * 90)


class_counts = (

    data[
        "Trend_Class"
    ]
    .value_counts()
)


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    class_counts.index,

    class_counts.values
)


plt.title(
    "Government Scheme Trend Class Distribution"
)


plt.xlabel(
    "Trend Class"
)


plt.ylabel(
    "Number of Schemes"
)


for bar, value in zip(
    bars,
    class_counts.values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 0.05,

        str(
            value
        ),

        ha="center",

        fontweight="bold"
    )


plt.tight_layout()


plt.savefig(

    PSO_CLASS_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_CLASS_GRAPH_PATH
)


# =============================================================================
# 44. CREATE PSO RESULT CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO RESULT CSV")
print("=" * 90)


pso_result_df = data[

    [

        "Scheme",

        "2018-19",

        "2019-20",

        "2020-21",

        "Total_Allocation",

        "Average_Allocation",

        "Minimum_Allocation",

        "Maximum_Allocation",

        "Std_Allocation",

        "Growth_2018_2019",

        "Growth_2019_2020",

        "Overall_Growth",

        "Growth_Percent_2018_2019",

        "Growth_Percent_2019_2020",

        "Overall_Growth_Percent",

        "Trend_Slope",

        "Coefficient_of_Variation",

        "Trend_Class"
    ]

].copy()


pso_result_df[
    "PSO_Selected_Feature_Count"
] = len(
    selected_features
)


pso_result_df[
    "PSO_Fitness"
] = (
    global_best_fitness
)


pso_result_df[
    "PSO_LOOCV_Accuracy"
] = (
    pso_cv_accuracy
)


pso_result_df.to_csv(

    PSO_RESULT_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    PSO_RESULT_CSV_PATH
)


print("\nPSO Result Data:")

print(
    pso_result_df
)


# =============================================================================
# 45. PSO RESULT GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO RESULT GRAPH")
print("=" * 90)


x_positions = np.arange(
    len(
        pso_result_df
    )
)


growth_values = (

    pso_result_df[
        "Overall_Growth_Percent"
    ]
)


plt.figure(
    figsize=(14, 7)
)


plt.bar(

    x_positions,

    growth_values
)


plt.axhline(
    y=0,
    linewidth=1
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Overall Growth (%)"
)


plt.title(
    "PSO Government Scheme Fund Growth Analysis"
)


plt.xticks(

    x_positions,

    pso_result_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


for i, value in enumerate(
    growth_values
):

    if value >= 0:

        vertical_alignment = (
            "bottom"
        )

    else:

        vertical_alignment = (
            "top"
        )


    plt.text(

        i,

        value,

        f"{value:.1f}%",

        ha="center",

        va=vertical_alignment,

        fontsize=8
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_RESULT_GRAPH_PATH
)


# =============================================================================
# 46. CREATE PSO PREDICTION CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO PREDICTION CSV")
print("=" * 90)


actual_labels = (
    label_encoder
    .inverse_transform(
        pso_actual
    )
)


predicted_labels = (
    label_encoder
    .inverse_transform(
        pso_predicted
    )
)


pso_prediction_df = pd.DataFrame(

    {

        "Scheme":
            data[
                "Scheme"
            ],

        "Actual_Trend":
            actual_labels,

        "PSO_Predicted_Trend":
            predicted_labels,

        "Correct_Prediction":
            (
                actual_labels
                ==
                predicted_labels
            ),

        "Actual_Encoded":
            pso_actual,

        "PSO_Predicted_Encoded":
            pso_predicted
    }
)


pso_prediction_df.to_csv(

    PSO_PREDICTION_CSV_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    PSO_PREDICTION_CSV_PATH
)


print("\nPSO Prediction Data:")

print(
    pso_prediction_df
)


# =============================================================================
# 47. PSO PREDICTION GRAPH
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO PREDICTION GRAPH")
print("=" * 90)


x_positions = np.arange(
    len(
        pso_prediction_df
    )
)


width = 0.35


plt.figure(
    figsize=(14, 7)
)


plt.bar(

    x_positions
    -
    width / 2,

    pso_prediction_df[
        "Actual_Encoded"
    ],

    width,

    label="Actual"
)


plt.bar(

    x_positions
    +
    width / 2,

    pso_prediction_df[
        "PSO_Predicted_Encoded"
    ],

    width,

    label="PSO Predicted"
)


plt.xlabel(
    "Government Scheme"
)


plt.ylabel(
    "Trend Class"
)


plt.title(
    "PSO Actual vs Predicted Government Scheme Funding Trend"
)


plt.xticks(

    x_positions,

    pso_prediction_df[
        "Scheme"
    ],

    rotation=45,

    ha="right"
)


class_positions = np.arange(
    len(
        label_encoder.classes_
    )
)


plt.yticks(

    class_positions,

    label_encoder.classes_
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PSO_PREDICTION_GRAPH_PATH
)


# =============================================================================
# 48. MODEL COMPARISON CSV
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO MODEL COMPARISON CSV")
print("=" * 90)


pso_comparison_df = pd.DataFrame(

    {

        "Model": [

            "Baseline Random Forest",

            "PSO Optimized Random Forest"
        ],

        "Number_of_Features": [

            len(
                FEATURE_COLUMNS
            ),

            len(
                selected_features
            )
        ],

        "Training_Accuracy": [

            baseline_training_accuracy,

            pso_training_accuracy
        ],

        "LOOCV_Accuracy": [

            baseline_cv_accuracy,

            pso_cv_accuracy
        ]
    }
)


pso_comparison_df.to_csv(

    PSO_MODEL_COMPARISON_PATH,

    index=False,

    encoding="utf-8-sig"
)


print(
    "\nSaved:",
    PSO_MODEL_COMPARISON_PATH
)


print("\nModel Comparison:")

print(
    pso_comparison_df
)


# =============================================================================
# 49. CONFUSION MATRIX
# =============================================================================

print("\n" + "=" * 90)
print("PSO CONFUSION MATRIX")
print("=" * 90)


pso_cm = confusion_matrix(

    pso_actual,

    pso_predicted,

    labels=np.arange(
        len(
            label_encoder.classes_
        )
    )
)


print(
    pso_cm
)


# =============================================================================
# 50. CLASSIFICATION REPORT
# =============================================================================

print("\n" + "=" * 90)
print("PSO CLASSIFICATION REPORT")
print("=" * 90)


print(

    classification_report(

        pso_actual,

        pso_predicted,

        labels=np.arange(
            len(
                label_encoder.classes_
            )
        ),

        target_names=
            label_encoder.classes_,

        zero_division=0
    )
)


# =============================================================================
# 51. CREATE PSO YAML CONFIG
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO CONFIGURATION")
print("=" * 90)


pso_config = {

    "project": {

        "name":
            "AI-Based Government Scheme Fund Allocation Trend Analysis",

        "optimization":
            "Particle Swarm Optimization",

        "algorithm":
            "Binary Particle Swarm Optimization",

        "classifier":
            "Random Forest"
    },


    "dataset": {

        "filename":
            "RS_Session_256_AU_3172_C.csv",

        "path":
            CSV_PATH,

        "rows":
            int(
                len(data)
            ),

        "financial_years":
            YEAR_COLUMNS
    },


    "pso_parameters": {

        "number_of_particles":
            PSO_NUM_PARTICLES,

        "iterations":
            PSO_ITERATIONS,

        "inertia_weight":
            PSO_INERTIA_WEIGHT,

        "cognitive_coefficient":
            PSO_COGNITIVE_COEFFICIENT,

        "social_coefficient":
            PSO_SOCIAL_COEFFICIENT
    },


    "features": {

        "total_features":
            len(
                FEATURE_COLUMNS
            ),

        "selected_feature_count":
            len(
                selected_features
            ),

        "selected_features":
            selected_features
    },


    "performance": {

        "best_pso_fitness":
            float(
                global_best_fitness
            ),

        "training_accuracy":
            float(
                pso_training_accuracy
            ),

        "loocv_accuracy":
            float(
                pso_cv_accuracy
            ),

        "baseline_training_accuracy":
            float(
                baseline_training_accuracy
            ),

        "baseline_loocv_accuracy":
            float(
                baseline_cv_accuracy
            )
    }
}


with open(

    PSO_CONFIG_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        pso_config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nSaved:",
    PSO_CONFIG_PATH
)


# =============================================================================
# 52. CREATE PSO JSON METADATA
# =============================================================================

print("\n" + "=" * 90)
print("CREATING PSO METADATA")
print("=" * 90)


pso_metadata = {

    "project_name":
        "AI-Based Government Scheme Fund Allocation Trend Analysis",

    "optimization_method":
        "Particle Swarm Optimization",

    "pso_variant":
        "Binary Particle Swarm Optimization",

    "classifier":
        "Random Forest",

    "created_at":
        datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),

    "dataset": {

        "filename":
            os.path.basename(
                CSV_PATH
            ),

        "rows":
            int(
                len(data)
            ),

        "financial_years":
            YEAR_COLUMNS
    },


    "pso_parameters": {

        "number_of_particles":
            int(
                PSO_NUM_PARTICLES
            ),

        "iterations":
            int(
                PSO_ITERATIONS
            ),

        "inertia_weight":
            float(
                PSO_INERTIA_WEIGHT
            ),

        "cognitive_coefficient":
            float(
                PSO_COGNITIVE_COEFFICIENT
            ),

        "social_coefficient":
            float(
                PSO_SOCIAL_COEFFICIENT
            )
    },


    "all_features":
        FEATURE_COLUMNS,


    "selected_features":
        selected_features,


    "selected_feature_count":
        int(
            len(
                selected_features
            )
        ),


    "best_particle":
        best_particle.tolist(),


    "best_fitness":
        float(
            global_best_fitness
        ),


    "training_accuracy":
        float(
            pso_training_accuracy
        ),


    "loocv_accuracy":
        float(
            pso_cv_accuracy
        ),


    "baseline_training_accuracy":
        float(
            baseline_training_accuracy
        ),


    "baseline_loocv_accuracy":
        float(
            baseline_cv_accuracy
        ),


    "classes":
        label_encoder.classes_.tolist(),


    "class_mapping": {

        str(index):
            str(class_name)

        for index, class_name
        in enumerate(
            label_encoder.classes_
        )
    },


    "class_distribution": {

        str(key):
            int(value)

        for key, value
        in (
            data[
                "Trend_Class"
            ]
            .value_counts()
            .items()
        )
    },


    "output_prefix":
        "pso_"
}


with open(

    PSO_METADATA_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        pso_metadata,

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nSaved:",
    PSO_METADATA_PATH
)


# =============================================================================
# 53. CALCULATE PSO IMPROVEMENT
# =============================================================================

accuracy_difference = (

    pso_cv_accuracy

    -

    baseline_cv_accuracy

) * 100


feature_reduction = (

    1

    -

    (
        len(
            selected_features
        )

        /

        len(
            FEATURE_COLUMNS
        )
    )

) * 100


print("\n" + "=" * 90)
print("PSO PERFORMANCE IMPROVEMENT")
print("=" * 90)


print(
    "\nBaseline LOOCV Accuracy:",
    f"{baseline_cv_accuracy * 100:.2f}%"
)


print(
    "PSO LOOCV Accuracy:",
    f"{pso_cv_accuracy * 100:.2f}%"
)


print(
    "Accuracy Difference:",
    f"{accuracy_difference:+.2f} percentage points"
)


print(
    "\nOriginal Features:",
    len(
        FEATURE_COLUMNS
    )
)


print(
    "PSO Selected Features:",
    len(
        selected_features
    )
)


print(
    "Feature Reduction:",
    f"{feature_reduction:.2f}%"
)


# =============================================================================
# 54. VERIFY ALL GENERATED FILES
# =============================================================================

print("\n" + "=" * 90)
print("VERIFYING PSO OUTPUT FILES")
print("=" * 90)


generated_files = [

    PSO_MODEL_PATH,

    PSO_CONFIG_PATH,

    PSO_METADATA_PATH,

    PSO_ACCURACY_GRAPH_PATH,

    PSO_HEATMAP_PATH,

    PSO_COMPARISON_GRAPH_PATH,

    PSO_FEATURE_GRAPH_PATH,

    PSO_CLASS_GRAPH_PATH,

    PSO_RESULT_CSV_PATH,

    PSO_RESULT_GRAPH_PATH,

    PSO_PREDICTION_CSV_PATH,

    PSO_PREDICTION_GRAPH_PATH,

    PSO_MODEL_COMPARISON_PATH,

    PSO_SELECTED_FEATURES_PATH
]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        file_size = (

            os.path.getsize(
                file_path
            )

            /

            1024
        )


        print(

            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({file_size:.2f} KB)"
        )


    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(file_path)}"
        )


# =============================================================================
# 55. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("PSO FINAL SUMMARY")
print("=" * 90)


print(
    "\nDataset Rows:",
    len(
        data
    )
)


print(
    "Original Feature Count:",
    len(
        FEATURE_COLUMNS
    )
)


print(
    "PSO Selected Feature Count:",
    len(
        selected_features
    )
)


print("\nPSO Selected Features:")


for feature in selected_features:

    print(
        "-",
        feature
    )


print(
    "\nBest PSO Fitness:",
    f"{global_best_fitness:.4f}"
)


print(
    "\nBaseline Training Accuracy:",
    f"{baseline_training_accuracy * 100:.2f}%"
)


print(
    "PSO Training Accuracy:",
    f"{pso_training_accuracy * 100:.2f}%"
)


print(
    "\nBaseline LOOCV Accuracy:",
    f"{baseline_cv_accuracy * 100:.2f}%"
)


print(
    "PSO LOOCV Accuracy:",
    f"{pso_cv_accuracy * 100:.2f}%"
)


print(
    "PSO Accuracy Difference:",
    f"{accuracy_difference:+.2f} percentage points"
)


print(
    "Feature Reduction:",
    f"{feature_reduction:.2f}%"
)


print("\nGenerated PSO Files:")


for file_path in generated_files:

    print(
        "-",
        os.path.basename(
            file_path
        )
    )


print("\n" + "=" * 90)
print("PSO PROCESS COMPLETED SUCCESSFULLY")
print("=" * 90)
```
